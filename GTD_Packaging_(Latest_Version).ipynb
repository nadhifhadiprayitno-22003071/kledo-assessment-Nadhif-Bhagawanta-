```python
# 3D Packing Optimizer Version 1.9 — Adds direct input from an existing Colab DataFrame
import plotly.graph_objects as go
import plotly.io as pio
from collections import Counter
from datetime import datetime
from pathlib import Path
import os
import re
import shutil
import tempfile
import traceback
import pandas as pd

pio.renderers.default = "colab"


# ============================================================
# DATA CLASSES
# ============================================================

class Box3D:
    """
    Represents an item box with dimensions, weight, and stacking permission.

    stackable:
    - False / unstackable: must be loaded into floor-only trucks.
    - True / stackable: loaded separately into stackable-only trucks, where
      vertical stacking is allowed.
    """

    def __init__(self, name, width, depth, height, weight, stackable=False):
        self.name = name
        self.width = width
        self.depth = depth
        self.height = height
        self.weight = weight
        self.stackable = stackable

    def volume(self):
        return self.width * self.depth * self.height


class Placement3D:
    """
    Stores the final position and orientation of a packed item box.
    """

    def __init__(self, box, x, y, z, width, depth, height, rotated=False):
        self.box = box
        self.x = x
        self.y = y
        self.z = z
        self.width = width
        self.depth = depth
        self.height = height
        self.rotated = rotated


class MainBox3D:
    """
    Represents a reusable main 3D container with:
    - dimensions,
    - score/cost,
    - maximum weight capacity.
    """

    def __init__(self, name, width, depth, height, max_weight):
        self.name = name
        self.width = width
        self.depth = depth
        self.height = height
        self.max_weight = max_weight

    def volume(self):
        return self.width * self.depth * self.height


# ============================================================
# BASIC PACKING HELPERS
# ============================================================

def volume(space):
    """
    Calculates the volume of a free-space in 3D.
    """
    return space["w"] * space["d"] * space["h"]


def get_orientations(box, allow_rotation=True):
    """
    Returns possible orientations of a 3D box.
    Width, depth, and height can be rearranged if rotation is allowed.
    """
    if not allow_rotation:
        # Original dimensions (width, depth, height, not rotated)
        return [(box.width, box.depth, box.height, False)]

    dims = [
        (box.width, box.depth, box.height),    # Original
        (box.width, box.height, box.depth),    # W H D
        (box.depth, box.width, box.height),    # D W H
        (box.depth, box.height, box.width),    # D H W
        (box.height, box.width, box.depth),    # H W D
        (box.height, box.depth, box.width),    # H D W
    ]

    unique_dims = []
    for dim in dims:
        if dim not in unique_dims:
            unique_dims.append(dim)

    orientations = []
    for w, d, h in unique_dims:
        # A box is considered "rotated" if its orientation is different from its original definition.
        rotated = (w, d, h) != (box.width, box.depth, box.height)
        orientations.append((w, d, h, rotated))

    return orientations


def can_fit_3d(width, depth, height, space):
    """
    Checks whether a box of given dimensions fits in a 3D space.
    """
    return (
        width <= space["w"] and
        depth <= space["d"] and
        height <= space["h"]
    )


def parse_stackable_value(value):
    """
    Converts manual/CSV stackable values into True/False.

    Accepted stackable values:
    - yes, y, true, t, 1, stackable

    Accepted unstackable values:
    - no, n, false, f, 0, unstackable, non-stackable, not stackable
    """
    if isinstance(value, bool):
        return value

    value_text = str(value).strip().lower()

    stackable_values = {
        "yes", "y", "true", "t", "1", "stackable"
    }

    unstackable_values = {
        "no", "n", "false", "f", "0",
        "unstackable", "non-stackable", "non stackable",
        "not stackable"
    }

    if value_text in stackable_values:
        return True

    if value_text in unstackable_values:
        return False

    raise ValueError(
        "Stackable must be yes/no, true/false, 1/0, "
        "stackable, or unstackable."
    )


def split_space_3d_floor_only(
    space,
    used_width,
    used_depth,
    item_gap
):
    """
    Splits the remaining floor area after placing an item.

    No space is created above the item.
    Every free space remains at z = 0.

    The split covers the entire remaining floor without overlap.
    """

    new_spaces = []

    occupied_width = min(
        used_width + item_gap,
        space["w"]
    )

    occupied_depth = min(
        used_depth + item_gap,
        space["d"]
    )

    # Right space uses the full remaining depth.
    right_space = {
        "x": space["x"] + occupied_width,
        "y": space["y"],
        "z": 0.0,
        "w": space["w"] - occupied_width,
        "d": space["d"],
        "h": space["h"]
    }

    # Front space only uses the width occupied by the item.
    # This prevents overlap with right_space.
    front_space = {
        "x": space["x"],
        "y": space["y"] + occupied_depth,
        "z": 0.0,
        "w": occupied_width,
        "d": space["d"] - occupied_depth,
        "h": space["h"]
    }

    for new_space in [right_space, front_space]:
        if (
            new_space["w"] > 0
            and new_space["d"] > 0
            and new_space["h"] > 0
        ):
            new_spaces.append(new_space)

    return new_spaces


def split_space_3d_with_stacking(
    space,
    used_width,
    used_depth,
    used_height,
    item_gap
):
    """
    Splits remaining 3D space after placing an item.

    This version is only used for stackable-only trucks.

    It creates:
    - right space,
    - front space,
    - upper space.

    The upper space allows another stackable item to be placed above the
    current stackable item.
    """

    new_spaces = []

    occupied_width = min(
        used_width + item_gap,
        space["w"]
    )

    occupied_depth = min(
        used_depth + item_gap,
        space["d"]
    )

    occupied_height = min(
        used_height + item_gap,
        space["h"]
    )

    right_space = {
        "x": space["x"] + occupied_width,
        "y": space["y"],
        "z": space["z"],
        "w": space["w"] - occupied_width,
        "d": space["d"],
        "h": space["h"]
    }

    front_space = {
        "x": space["x"],
        "y": space["y"] + occupied_depth,
        "z": space["z"],
        "w": occupied_width,
        "d": space["d"] - occupied_depth,
        "h": space["h"]
    }

    upper_space = {
        "x": space["x"],
        "y": space["y"],
        "z": space["z"] + occupied_height,
        "w": used_width,
        "d": used_depth,
        "h": space["h"] - occupied_height
    }

    for new_space in [right_space, front_space, upper_space]:
        if (
            new_space["w"] > 0
            and new_space["d"] > 0
            and new_space["h"] > 0
        ):
            new_spaces.append(new_space)

    return new_spaces


# ============================================================
# SINGLE MAIN-BOX PACKING ALGORITHM (3D)
# ============================================================

def pack_boxes_3d(
    container_width,
    container_depth,
    container_height,
    boxes,
    max_weight,
    allow_rotation=True,
    item_gap=0.0,
    container_gap=0.0,
    allow_stacking=False
):
    """
    Packs item boxes into one truck.

    If allow_stacking is False:
    - strict floor loading is used,
    - every item is placed at z = 0,
    - no item may be placed above another item.

    If allow_stacking is True:
    - upper spaces may be created,
    - items may be placed at z > 0,
    - this should only be used for stackable-only item groups.

    Weight, rotation, item gaps, and container gaps remain active.
    """

    sorted_boxes = boxes.copy()

    # Larger floor footprints are attempted first.
    sorted_boxes.sort(
        key=lambda box: (
            box.width * box.depth,
            box.volume()
        ),
        reverse=True
    )

    # Container gap applies to the side walls and the ceiling.
    # The floor itself stays at z = 0.
    effective_width = container_width - (2 * container_gap)
    effective_depth = container_depth - (2 * container_gap)
    effective_height = container_height - container_gap

    if (
        effective_width <= 0
        or effective_depth <= 0
        or effective_height <= 0
    ):
        return [], boxes.copy(), 0.0

    # Every free space represents an available floor region.
    free_spaces = [
        {
            "x": container_gap,
            "y": container_gap,
            "z": 0.0,
            "w": effective_width,
            "d": effective_depth,
            "h": effective_height
        }
    ]

    placements = []
    unpacked = []
    current_weight = 0.0

    for box in sorted_boxes:
        best_space_index = None
        best_orientation = None
        best_waste = None

        # Weight-capacity constraint.
        if current_weight + box.weight > max_weight:
            unpacked.append(box)
            continue

        for index, space in enumerate(free_spaces):

            # In floor-only mode, free spaces must stay at z = 0.
            # In stacking mode, z > 0 spaces are allowed.
            if not allow_stacking and space["z"] != 0:
                continue

            for width, depth, height, rotated in get_orientations(
                box,
                allow_rotation
            ):
                # The item must fit in floor width/depth and truck height.
                if can_fit_3d(
                    width,
                    depth,
                    height,
                    space
                ):
                    # Floor-only mode compares unused floor area.
                    # Stacking mode compares unused volume so upper spaces can
                    # be used more naturally.
                    if allow_stacking:
                        waste = (
                            space["w"] * space["d"] * space["h"]
                        ) - (
                            width * depth * height
                        )
                    else:
                        waste = (
                            space["w"] * space["d"]
                        ) - (
                            width * depth
                        )

                    if (
                        best_waste is None
                        or waste < best_waste
                    ):
                        best_waste = waste
                        best_space_index = index
                        best_orientation = (
                            width,
                            depth,
                            height,
                            rotated
                        )

        # No suitable floor region exists in this truck.
        if best_space_index is None:
            unpacked.append(box)
            continue

        selected_space = free_spaces.pop(best_space_index)

        used_width, used_depth, used_height, rotated = (
            best_orientation
        )

        placement = Placement3D(
            box=box,
            x=selected_space["x"],
            y=selected_space["y"],

            # Floor-only items stay at z=0. Stackable-only trucks may use z>0.
            z=selected_space["z"] if allow_stacking else 0.0,

            width=used_width,
            depth=used_depth,
            height=used_height,
            rotated=rotated
        )

        placements.append(placement)
        current_weight += box.weight

        if allow_stacking:
            # Stackable-only trucks may create right, front, and upper spaces.
            new_spaces = split_space_3d_with_stacking(
                selected_space,
                used_width,
                used_depth,
                used_height,
                item_gap
            )
        else:
            # Unstackable trucks only create floor spaces.
            new_spaces = split_space_3d_floor_only(
                selected_space,
                used_width,
                used_depth,
                item_gap
            )

        free_spaces.extend(new_spaces)

        if allow_stacking:
            # Prefer lower spaces first, then smaller volumes.
            free_spaces.sort(
                key=lambda space: (
                    space["z"],
                    space["w"] * space["d"] * space["h"]
                )
            )
        else:
            # Try smaller floor spaces first.
            free_spaces.sort(
                key=lambda space: space["w"] * space["d"]
            )

    return placements, unpacked, current_weight

def center_placements_3d(placements, container_width, container_depth, container_height, container_gap=0.0):
    """
    Centers a group of placements within the available space of a container.
    The available space considers the container_gap.
    This function modifies the placements in place.
    """
    if not placements:
        return placements

    # Calculate the bounding box of all placed items
    min_x = min(p.x for p in placements)
    min_y = min(p.y for p in placements)

    max_x = max(p.x + p.width for p in placements)
    max_y = max(p.y + p.depth for p in placements)

    group_width = max_x - min_x
    group_depth = max_y - min_y

    # Calculate effective container dimensions for centering
    effective_width = container_width - (2 * container_gap)
    effective_depth = container_depth - (2 * container_gap)

    # Calculate offsets to center the group
    # The new 'start' for the group should be container_gap + (effective_dim - group_dim) / 2
    # The offset to apply to each item's coordinate is (new_start - current_min)
    offset_x = container_gap + (effective_width - group_width) / 2 - min_x
    offset_y = container_gap + (effective_depth - group_depth) / 2 - min_y

    for p in placements:
        p.x += offset_x
        p.y += offset_y

    return placements

# ============================================================
# MULTIPLE MAIN-BOX SELECTION ALGORITHM (3D)
# ============================================================

def pack_multiple_main_boxes_3d(
    main_boxes_3d,
    item_boxes_3d,
    allow_rotation=True,
    item_gap=0.0,
    container_gap=0.0,
    center_items=False,
    allow_stacking=False,
    cargo_group_name="Cargo"
):
    """
    Repeatedly chooses and uses 3D main boxes until all possible item boxes
    have been packed.

    Main-box types may be used repeatedly.

    Selection logic:
    1. Try every available main-box type on the remaining item boxes.
    2. If a truck can pack all remaining items, choose the first such truck
       based on the truck list order.
    3. Otherwise, choose the truck that packs the most volume.
    4. Remove the packed item boxes.
    5. Repeat until all item boxes are packed or no main box can
       fit any remaining item.

    This is a greedy heuristic.
    """

    remaining_boxes = item_boxes_3d.copy()
    used_main_boxes = []
    # total_score = 0.0 # Score is no longer used
    usage_number = 1

    while remaining_boxes:
        candidates = []

        for main_box in main_boxes_3d:
            test_placements, test_unpacked, fitted_weight = pack_boxes_3d(
                main_box.width,
                main_box.depth,
                main_box.height,
                remaining_boxes,
                main_box.max_weight,
                allow_rotation,
                item_gap,
                container_gap,
                allow_stacking
            )

            fitted_count = len(test_placements)

            if fitted_count == 0:
                continue

            # Apply centering if requested
            if center_items:
                test_placements = center_placements_3d(
                    test_placements,
                    main_box.width,
                    main_box.depth,
                    main_box.height,
                    container_gap
                )

            fitted_floor_area = sum(
                placement.width * placement.depth
                for placement in test_placements
            )

            main_box_floor_area = (
                main_box.width * main_box.depth
            )

            floor_utilization = (
                fitted_floor_area / main_box_floor_area
            ) * 100

            fitted_volume = sum(
                p.width * p.depth * p.height
                for p in test_placements
            )

            volume_utilization = (fitted_volume / main_box.volume()) * 100
            weight_utilization = (fitted_weight / main_box.max_weight) * 100

            # Efficiency: packed item volume per score point (score is removed, so this is no longer calculated for selection)
            # efficiency = fitted_volume / main_box.score

            candidates.append({
                "main_box": main_box,
                "placements": test_placements,
                "unpacked": test_unpacked,
                "fitted_count": fitted_count,
                "fitted_volume": fitted_volume,
                "fitted_floor_area": fitted_floor_area, # Added fitted_floor_area
                "fitted_weight": fitted_weight,
                "floor_utilization": floor_utilization, # Added floor_utilization
                "volume_utilization": volume_utilization, # Renamed from 'utilization'
                "weight_utilization": weight_utilization
                # "efficiency": efficiency # Removed as score is no longer used for selection
            })

        if not candidates:
            break

        # Order-based logic: Prioritize the first truck that can pack everything
        complete_candidates = [c for c in candidates if len(c["unpacked"]) == 0]

        if complete_candidates:
            # Select the first complete candidate found based on the main_boxes order (CSV order)
            best_choice = complete_candidates[0]
        else:
            # Fallback to the one that fits the most volume, then fitted count as tie-breaker
            best_choice = max(
                candidates,
                key=lambda candidate: (
                    candidate["fitted_volume"],
                    candidate["fitted_count"]
                )
            )

        selected_main_box = best_choice["main_box"]

        used_main_boxes.append({
            "usage_number": usage_number,
            "cargo_group": cargo_group_name,
            "allow_stacking": allow_stacking,
            "main_box": selected_main_box,
            "placements": best_choice["placements"],
            "fitted_floor_area": best_choice["fitted_floor_area"],
            "fitted_volume": best_choice["fitted_volume"],
            "fitted_weight": best_choice["fitted_weight"],
            "floor_utilization": best_choice["floor_utilization"],
            "volume_utilization": best_choice["volume_utilization"],
            "weight_utilization": best_choice["weight_utilization"]
            # "score": selected_main_box.score # Score is no longer used
        })

        # total_score += selected_main_box.score # Score is no longer used
        remaining_boxes = best_choice["unpacked"]
        usage_number += 1

    # Removed total_score calculation since score is not used
    return used_main_boxes, remaining_boxes, 0.0 # Return 0.0 for total_score


def pack_separated_stackable_and_unstackable_items(
    main_boxes_3d,
    item_boxes_3d,
    allow_rotation=True,
    item_gap=0.0,
    container_gap=0.0,
    center_items=False
):
    """
    Separates items into two independent truck groups.

    1. Unstackable items:
       - loaded into their own trucks,
       - floor-only packing,
       - z always stays 0.

    2. Stackable items:
       - loaded into their own trucks,
       - stacking-enabled packing,
       - z may be greater than 0.

    The two groups are never mixed inside the same truck.
    """
    unstackable_boxes = [
        box for box in item_boxes_3d
        if not box.stackable
    ]

    stackable_boxes = [
        box for box in item_boxes_3d
        if box.stackable
    ]

    used_unstackable = []
    remaining_unstackable = []
    used_stackable = []
    remaining_stackable = []

    if unstackable_boxes:
        used_unstackable, remaining_unstackable, _ = pack_multiple_main_boxes_3d(
            main_boxes_3d=main_boxes_3d,
            item_boxes_3d=unstackable_boxes,
            allow_rotation=allow_rotation,
            item_gap=item_gap,
            container_gap=container_gap,
            center_items=center_items,
            allow_stacking=False,
            cargo_group_name="Unstackable"
        )

    if stackable_boxes:
        used_stackable, remaining_stackable, _ = pack_multiple_main_boxes_3d(
            main_boxes_3d=main_boxes_3d,
            item_boxes_3d=stackable_boxes,
            allow_rotation=allow_rotation,
            item_gap=item_gap,
            container_gap=container_gap,
            center_items=center_items,
            allow_stacking=True,
            cargo_group_name="Stackable"
        )

    used_all = used_unstackable + used_stackable

    # Renumber trucks after combining both groups.
    for index, result in enumerate(used_all, start=1):
        result["usage_number"] = index

    remaining_all = remaining_unstackable + remaining_stackable

    return used_all, remaining_all, 0.0


# ============================================================
# TEXT-BASED VISUALIZATION
# ============================================================

def print_3d_layers(container_width, container_depth, container_height, placements, main_box_name="Container"):
    """
    Prints the 3D container layer by layer.
    Each layer is a 2D map at a specific height z.
    Works best with whole number dimensions.
    """
    grid_width = int(container_width)
    grid_depth = int(container_depth)
    grid_height = int(container_height)

    grid = [
        [
            ["." for _ in range(grid_width)]
            for _ in range(grid_depth)
        ]
        for _ in range(grid_height)
    ]

    symbols = list("ABCDEFGHIJKLMNOPQRSTUVWXYZabcdefghijklmnopqrstuvwxyz0123456789")
    box_symbols = {}

    for index, placement in enumerate(placements):
        symbol = symbols[index] if index < len(symbols) else str(index + 1)
        box_symbols[placement.box.name] = symbol

        x_start = int(placement.x)
        y_start = int(placement.y)
        z_start = int(placement.z)

        # Iterate over the placed box's dimensions, clamped to container bounds.
        # This handles cases where gaps might push effective placement outside grid logic
        # (though ideally the packing algorithm prevents this for the box itself).
        for z in range(z_start, min(z_start + int(placement.height), grid_height)):
            for y in range(y_start, min(y_start + int(placement.depth), grid_depth)):
                for x in range(x_start, min(x_start + int(placement.width), grid_width)):
                    if 0 <= x < grid_width and 0 <= y < grid_depth and 0 <= z < grid_height:
                        grid[z][y][x] = symbol

    print(f"\n=== {main_box_name} 3D Container Layout by Height Layer ===")

    for z in range(grid_height):
        print(f"\nLayer z={z}")
        for y in range(grid_depth):
            print(" ".join(grid[z][y]))

    print("\nLegend:")
    for placement in placements:
        symbol = box_symbols[placement.box.name]
        print(
            f"{symbol} = {placement.box.name}: "
            f"{placement.width} x {placement.depth} x {placement.height}, "
            f"weight={placement.box.weight}, " # Added weight
            f"x={placement.x}, y={placement.y}, z={placement.z}, "
            f"rotated={placement.rotated}"
        )

    print(". = Empty space")


def create_box_mesh(x, y, z, w, d, h):
    """
    Creates the corner points and faces for a 3D rectangular box.
    This is used by Plotly Mesh3d.
    """
    vertices = [
        [x,     y,     z],        # 0
        [x + w, y,     z],        # 1
        [x + w, y + d, z],        # 2
        [x,     y + d, z],        # 3
        [x,     y,     z + h],    # 4
        [x + w, y,     z + h],    # 5
        [x + w, y + d, z + h],    # 6
        [x,     y + d, z + h],    # 7
    ]

    x_coords = [v[0] for v in vertices]
    y_coords = [v[1] for v in vertices]
    z_coords = [v[2] for v in vertices]

    # Define the 12 triangles that make up the 6 faces of the cube
    # Each face needs two triangles
    i = [0, 0, 0, 1, 1, 2, 4, 4, 4, 5, 5, 6]
    j = [1, 2, 4, 2, 5, 3, 5, 6, 7, 6, 1, 2]
    k = [2, 3, 5, 5, 6, 7, 6, 7, 0, 2, 0, 3]

    return x_coords, y_coords, z_coords, i, j, k


def visualize_3d_packing(
    container_width,
    container_depth,
    container_height,
    placements,
    title="3D Container Packing Visualization",
    show=True,
    camera=None
):
    """
    Builds a Plotly 3D figure.

    Parameters:
    - show=True: display the interactive figure in Colab/Jupyter.
    - show=False: build the same figure without displaying it, useful for PDF export.
    - camera: optional Plotly camera dictionary for a fixed PDF view.

    Returns:
        plotly.graph_objects.Figure
    """
    fig = go.Figure()

    colors = [
        "red", "green", "blue", "orange", "purple", "cyan",
        "magenta", "yellow", "brown", "pink", "lime", "teal",
        "gold", "navy", "coral", "olive"
    ]

    for index, placement in enumerate(placements):
        x = placement.x
        y = placement.y
        z = placement.z
        w = placement.width
        d = placement.depth
        h = placement.height

        x_coords, y_coords, z_coords, i, j, k = create_box_mesh(
            x, y, z, w, d, h
        )

        fig.add_trace(go.Mesh3d(
            x=x_coords,
            y=y_coords,
            z=z_coords,
            i=i,
            j=j,
            k=k,
            color=colors[index % len(colors)],
            opacity=0.75,
            name=placement.box.name,
            hovertext=(
                f"Item: {placement.box.name}<br>"
                f"Position: x={x:.2f}, y={y:.2f}, z={z:.2f}<br>"
                f"Size: {w:.2f} x {d:.2f} x {h:.2f}<br>"
                f"Original Size: "
                f"{placement.box.width:.2f} x "
                f"{placement.box.depth:.2f} x "
                f"{placement.box.height:.2f}<br>"
                f"Weight: {placement.box.weight:.2f}<br>"
                f"Rotated: {placement.rotated}"
            ),
            hoverinfo="text"
        ))

        add_box_outline(
            fig, x, y, z, w, d, h, placement.box.name
        )

    add_container_outline(
        fig,
        container_width,
        container_depth,
        container_height
    )

    scene_settings = dict(
        xaxis_title="Length",
        yaxis_title="Width",
        zaxis_title="Height",
        xaxis=dict(range=[0, container_width]),
        yaxis=dict(range=[0, container_depth]),
        zaxis=dict(range=[0, container_height]),
        aspectmode="data"
    )

    if camera is not None:
        scene_settings["camera"] = camera

    fig.update_layout(
        title=title,
        scene=scene_settings,
        width=950,
        height=750,
        showlegend=True,
        margin=dict(l=20, r=20, t=80, b=20)
    )

    if show:
        fig.show()

    return fig


def add_box_outline(fig, x, y, z, w, d, h, name):
    """
    Adds black edge lines around each placed box.
    """
    # Define corners of the box
    corners = {
        "000": (x,     y,     z),        # 0
        "100": (x + w, y,     z),        # 1
        "110": (x + w, y + d, z),        # 2
        "010": (x,     y + d, z),        # 3
        "001": (x,     y,     z + h),    # 4
        "101": (x + w, y,     z + h),    # 5
        "111": (x + w, y + d, z + h),    # 6
        "011": (x,     y + d, z + h),    # 7
    }

    # Define the 12 edges of a cube by connecting corners
    edges = [
        ("000", "100"), ("100", "110"), ("110", "010"), ("010", "000"), # Bottom face
        ("001", "101"), ("101", "111"), ("111", "011"), ("011", "001"), # Top face
        ("000", "001"), ("100", "101"), ("110", "111"), ("010", "011")  # Vertical edges
    ]

    edge_x = []
    edge_y = []
    edge_z = []

    for start, end in edges:
        x1, y1, z1 = corners[start]
        x2, y2, z2 = corners[end]

        edge_x.extend([x1, x2, None]) # Add None to break the line for scatter3d
        edge_y.extend([y1, y2, None])
        edge_z.extend([z1, z2, None])

    fig.add_trace(go.Scatter3d(
        x=edge_x,
        y=edge_y,
        z=edge_z,
        mode="lines",
        line=dict(width=3, color="black"),
        name=f"{name} outline",
        showlegend=False
    ))


def add_container_outline(fig, container_width, container_depth, container_height):
    """
    Adds an outline showing the total container/platform boundary.
    """
    x = 0
    y = 0
    z = 0
    w = container_width
    d = container_depth
    h = container_height

    corners = {
        "000": (x,     y,     z),
        "100": (x + w, y,     z),
        "110": (x + w, y + d, z),
        "010": (x,     y + d, z),
        "001": (x,     y,     z + h),
        "101": (x + w, y,     z + h),
        "111": (x + w, y + d, z + h),
        "011": (x,     y + d, z + h),
    }

    edges = [
        ("000", "100"), ("100", "110"), ("110", "010"), ("010", "000"),
        ("001", "101"), ("101", "111"), ("111", "011"), ("011", "001"),
        ("000", "001"), ("100", "101"), ("110", "111"), ("010", "011")
    ]

    edge_x = []
    edge_y = []
    edge_z = []

    for start, end in edges:
        x1, y1, z1 = corners[start]
        x2, y2, z2 = corners[end]

        edge_x.extend([x1, x2, None])
        edge_y.extend([y1, y2, None])
        edge_z.extend([z1, z2, None])

    fig.add_trace(go.Scatter3d(
        x=edge_x,
        y=edge_y,
        z=edge_z,
        mode="lines",
        line=dict(width=7, color="black"),
        name="Container Boundary"
    ))


def visualize_multiple_main_boxes_3d(used_main_boxes_3d):
    """
    Displays one interactive Plotly visualization for every truck used.

    Returns:
        list[plotly.graph_objects.Figure]
    """
    figures = []

    for result in used_main_boxes_3d:
        main_box = result["main_box"]

        cargo_group = result.get("cargo_group", "Cargo")
        stacking_label = (
            "Stacking Enabled"
            if result.get("allow_stacking", False)
            else "Floor Only"
        )

        title = (
            f"Main Box {result['usage_number']} "
            f"[{cargo_group} - {stacking_label}]: "
            f"{main_box.name} "
            f"({main_box.width} x "
            f"{main_box.depth} x "
            f"{main_box.height} cm)"
            f"<br>Boxes: {len(result['placements'])} | "
            f"Floor Utilization: "
            f"{result['floor_utilization']:.2f}% | "
            f"Weight Utilization: "
            f"{result['weight_utilization']:.2f}%"
        )

        figure = visualize_3d_packing(
            main_box.width,
            main_box.depth,
            main_box.height,
            result["placements"],
            title=title,
            show=True
        )
        figures.append(figure)

    return figures



# ============================================================
# INTERACTIVE HTML REPORT EXPORT
# ============================================================

def create_packing_report_html(
    used_main_boxes_3d,
    remaining_boxes,
    total_item_count,
    packed_count,
    item_gap,
    container_gap,
    center_items,
    output_filename="packing_report.html"
):
    """Creates a standalone interactive HTML packing report."""
    from html import escape

    output_path = Path(output_filename).resolve()
    truck_counts = Counter(
        (
            result.get("cargo_group", "Cargo"),
            result["main_box"].name,
            "Stacking Enabled" if result.get("allow_stacking", False) else "Floor Only"
        )
        for result in used_main_boxes_3d
    )

    parts = ["""<!DOCTYPE html>
<html lang="en"><head><meta charset="UTF-8">
<meta name="viewport" content="width=device-width, initial-scale=1.0">
<title>3D Packing Optimizer Report</title>
<style>
body{margin:0;padding:30px;background:#f4f6f8;color:#222;font-family:Arial,sans-serif}
.wrap{max-width:1400px;margin:auto}.card{background:white;border-radius:10px;padding:20px;margin-bottom:25px;box-shadow:0 2px 8px rgba(0,0,0,.08)}
h1{text-align:center}table{width:100%;border-collapse:collapse;margin-top:10px}th,td{border:1px solid #bbb;padding:8px;text-align:left;vertical-align:top}th{background:#d9eaf7}tr:nth-child(even){background:#f7f7f7}.help{background:#eef5ff;border-left:4px solid #3b78b4;padding:12px;margin-bottom:20px}.plot{min-height:720px}.warn{color:#b14b00;font-weight:bold}.good{color:#167c31;font-weight:bold}
</style></head><body><div class="wrap">"""]
    parts.append("<h1>Multi-Main-Box 3D Packing Report</h1>")
    parts.append(
        '<p style="text-align:center;color:#666">'
        f'Generated: {datetime.now().strftime("%d %B %Y, %H:%M")}</p>'
    )
    parts.append(
        '<div class="help">Drag to rotate. Use the mouse wheel to zoom. '
        'Hold Shift and drag to pan. Hover over cargo for details.</div>'
    )

    status = "All item boxes were packed." if not remaining_boxes else "Some item boxes could not be packed."
    status_class = "good" if not remaining_boxes else "warn"
    parts.append(f"""<div class="card"><h2>Summary Result</h2><table>
<tr><th>Summary</th><th>Result</th></tr>
<tr><td>Total item boxes entered</td><td>{total_item_count}</td></tr>
<tr><td>Total item boxes packed</td><td>{packed_count}</td></tr>
<tr><td>Total item boxes not packed</td><td>{len(remaining_boxes)}</td></tr>
<tr><td>Total main boxes used</td><td>{len(used_main_boxes_3d)}</td></tr>
<tr><td>Container clearance</td><td>{container_gap:.2f} cm</td></tr>
<tr><td>Item-to-item gap</td><td>{item_gap:.2f} cm</td></tr>
<tr><td>Center items</td><td>{center_items}</td></tr>
<tr><td>Packing status</td><td class="{status_class}">{status}</td></tr></table></div>""")

    parts.append('<div class="card"><h2>Truck Types Used</h2><table><tr><th>Amount</th><th>Truck type</th><th>Cargo group</th><th>Stacking mode</th></tr>')
    for (cargo_group, truck_name, stacking_label), count in truck_counts.items():
        parts.append(
            f'<tr><td>{count}</td><td>{escape(str(truck_name))}</td>'
            f'<td>{escape(str(cargo_group))}</td><td>{escape(stacking_label)}</td></tr>'
        )
    parts.append('</table></div>')

    if remaining_boxes:
        parts.append('<div class="card"><h2>Item Boxes That Could Not Fit</h2><table><tr><th>Item type</th><th>Length × Width × Height</th><th>Weight/item</th><th>Stacking</th><th>Amount</th></tr>')
        for (name, width, depth, height, weight, stackable), count in group_unpacked_boxes(remaining_boxes).items():
            parts.append(
                f'<tr><td>{escape(str(name))}</td>'
                f'<td>{width:.1f} × {depth:.1f} × {height:.1f} cm</td>'
                f'<td>{weight:.1f} kg</td>'
                f'<td>{"Stackable" if stackable else "Unstackable"}</td>'
                f'<td>{count}</td></tr>'
            )
        parts.append('</table></div>')

    include_js = True
    for result in used_main_boxes_3d:
        main_box = result["main_box"]
        placements = result["placements"]
        usage_number = result["usage_number"]
        cargo_group = result.get("cargo_group", "Cargo")
        stacking_label = "Stacking Enabled" if result.get("allow_stacking", False) else "Floor Only"

        figure = visualize_3d_packing(
            main_box.width,
            main_box.depth,
            main_box.height,
            placements,
            title=(
                f"Truck {usage_number}: {main_box.name}<br>"
                f"{cargo_group} — {stacking_label}<br>"
                f"Boxes: {len(placements)} | "
                f"Floor utilization: {result['floor_utilization']:.2f}% | "
                f"Weight utilization: {result['weight_utilization']:.2f}%"
            ),
            show=False
        )
        figure.update_layout(autosize=True, width=None, height=720)
        plot_html = pio.to_html(
            figure,
            full_html=False,
            include_plotlyjs=True if include_js else False,
            config={"responsive": True, "displaylogo": False, "scrollZoom": True}
        )
        include_js = False

        parts.append(f"""<div class="card"><h2>Used Main Box {usage_number}: {escape(main_box.name)}</h2><table>
<tr><th>Truck detail</th><th>Value</th></tr>
<tr><td>Cargo group</td><td>{escape(str(cargo_group))}</td></tr>
<tr><td>Stacking mode</td><td>{escape(stacking_label)}</td></tr>
<tr><td>Cargo dimensions</td><td>{main_box.width:.1f} × {main_box.depth:.1f} × {main_box.height:.1f} cm</td></tr>
<tr><td>Maximum weight</td><td>{main_box.max_weight:.2f} kg</td></tr>
<tr><td>Boxes packed</td><td>{len(placements)}</td></tr>
<tr><td>Floor utilization</td><td>{result['floor_utilization']:.2f}%</td></tr>
<tr><td>Weight utilization</td><td>{result['weight_utilization']:.2f}%</td></tr>
<tr><td>Volume utilization</td><td>{result['volume_utilization']:.2f}%</td></tr>
</table><div class="plot">{plot_html}</div>
<h3>Box Placements</h3><table><tr><th>Item</th><th>x</th><th>y</th><th>z</th><th>L</th><th>W</th><th>H</th><th>kg</th><th>Rotated</th><th>Stackable</th></tr>""")
        for placement in placements:
            parts.append(
                f'<tr><td>{escape(placement.box.name)}</td>'
                f'<td>{placement.x:.1f}</td><td>{placement.y:.1f}</td><td>{placement.z:.1f}</td>'
                f'<td>{placement.width:.1f}</td><td>{placement.depth:.1f}</td><td>{placement.height:.1f}</td>'
                f'<td>{placement.box.weight:.1f}</td>'
                f'<td>{"Yes" if placement.rotated else "No"}</td>'
                f'<td>{"Yes" if placement.box.stackable else "No"}</td></tr>'
            )
        parts.append('</table></div>')

    parts.append('</div></body></html>')
    output_path.write_text(''.join(parts), encoding='utf-8')
    if not output_path.exists() or output_path.stat().st_size == 0:
        raise RuntimeError("The HTML report was not created correctly.")
    return str(output_path)


# ============================================================
# ROBUST STATIC IMAGE EXPORT HELPERS
# ============================================================

def configure_kaleido_browser():
    """
    Tries to locate Chrome/Chromium and exposes it to Kaleido through
    the BROWSER_PATH environment variable.

    The commands below should still be run once in Google Colab:

        !pip install -q -U plotly kaleido reportlab
        !plotly_get_chrome -y

    Returns:
        str | None: detected browser path, or None if no browser was found.
    """
    existing = os.environ.get("BROWSER_PATH")
    if existing and Path(existing).exists():
        return existing

    executable_names = [
        "google-chrome",
        "google-chrome-stable",
        "chromium",
        "chromium-browser",
        "chrome",
    ]

    for executable_name in executable_names:
        detected = shutil.which(executable_name)
        if detected:
            os.environ["BROWSER_PATH"] = detected
            return detected

    common_paths = [
        "/usr/bin/google-chrome",
        "/usr/bin/google-chrome-stable",
        "/usr/bin/chromium",
        "/usr/bin/chromium-browser",
        "/opt/google/chrome/chrome",
    ]

    for candidate in common_paths:
        if Path(candidate).exists():
            os.environ["BROWSER_PATH"] = candidate
            return candidate

    return None


def export_plotly_figure_to_png(
    figure,
    image_path,
    width=1300,
    height=850,
    scale=1
):
    """
    Exports a Plotly figure to PNG using Kaleido.

    This function:
    - explicitly detects Chrome/Chromium,
    - uses plotly.io.to_image instead of hiding the real exception,
    - verifies that the resulting PNG is non-empty.

    Raises:
        RuntimeError: with the real Plotly/Kaleido error.
    """
    browser_path = configure_kaleido_browser()

    try:
        image_bytes = pio.to_image(
            figure,
            format="png",
            width=width,
            height=height,
            scale=scale
        )

        if not image_bytes:
            raise RuntimeError("Kaleido returned empty image data.")

        image_path = Path(image_path)
        image_path.write_bytes(image_bytes)

        if not image_path.exists() or image_path.stat().st_size == 0:
            raise RuntimeError("The PNG file was not created correctly.")

        return str(image_path)

    except Exception as exc:
        browser_message = (
            browser_path
            if browser_path
            else "Chrome/Chromium was not detected"
        )

        raise RuntimeError(
            "Plotly/Kaleido static image export failed.\n"
            f"Detected browser: {browser_message}\n"
            f"Actual error type: {type(exc).__name__}\n"
            f"Actual error message: {exc}"
        ) from exc


def save_matplotlib_truck_image(result, image_path):
    """
    Fallback PDF image renderer.

    If Plotly/Kaleido still fails in Colab, this creates a static 3D truck
    image with Matplotlib. The interactive Plotly figures remain unchanged
    in the notebook; only the PDF snapshot uses this fallback.
    """
    try:
        import matplotlib.pyplot as plt
        from matplotlib.patches import Patch
    except ImportError as exc:
        raise RuntimeError(
            "Matplotlib fallback is unavailable. "
            "Install it with: !pip install -q matplotlib"
        ) from exc

    main_box = result["main_box"]
    placements = result["placements"]

    fig = plt.figure(figsize=(12, 8))
    ax = fig.add_subplot(111, projection="3d")

    color_cycle = list(plt.cm.tab20.colors)
    legend_handles = []

    for index, placement in enumerate(placements):
        color = color_cycle[index % len(color_cycle)]

        ax.bar3d(
            placement.x,
            placement.y,
            placement.z,
            placement.width,
            placement.depth,
            placement.height,
            color=color,
            edgecolor="black",
            linewidth=0.6,
            alpha=0.75,
            shade=True
        )

        legend_handles.append(
            Patch(
                facecolor=color,
                edgecolor="black",
                label=placement.box.name
            )
        )

    # Container boundary.
    x0, y0, z0 = 0.0, 0.0, 0.0
    x1 = main_box.width
    y1 = main_box.depth
    z1 = main_box.height

    corners = [
        (x0, y0, z0), (x1, y0, z0), (x1, y1, z0), (x0, y1, z0),
        (x0, y0, z1), (x1, y0, z1), (x1, y1, z1), (x0, y1, z1),
    ]

    edges = [
        (0, 1), (1, 2), (2, 3), (3, 0),
        (4, 5), (5, 6), (6, 7), (7, 4),
        (0, 4), (1, 5), (2, 6), (3, 7),
    ]

    for start_index, end_index in edges:
        start = corners[start_index]
        end = corners[end_index]

        ax.plot(
            [start[0], end[0]],
            [start[1], end[1]],
            [start[2], end[2]],
            color="black",
            linewidth=1.5
        )

    ax.set_xlim(0, main_box.width)
    ax.set_ylim(0, main_box.depth)
    ax.set_zlim(0, main_box.height)

    ax.set_xlabel("Length (cm)")
    ax.set_ylabel("Width (cm)")
    ax.set_zlabel("Height (cm)")

    ax.set_title(
        f"Truck {result['usage_number']}: {main_box.name}\n"
        f"{result.get('cargo_group', 'Cargo')} - "
        f"{'Stacking Enabled' if result.get('allow_stacking', False) else 'Floor Only'} | "
        f"Boxes: {len(placements)} | "
        f"Floor utilization: {result['floor_utilization']:.2f}% | "
        f"Weight utilization: {result['weight_utilization']:.2f}%"
    )

    ax.view_init(elev=24, azim=-55)

    # Keep the legend manageable for large runs.
    if legend_handles:
        maximum_legend_items = 16
        handles_to_show = legend_handles[:maximum_legend_items]

        if len(legend_handles) > maximum_legend_items:
            handles_to_show.append(
                Patch(
                    facecolor="white",
                    edgecolor="white",
                    label=(
                        f"... and "
                        f"{len(legend_handles) - maximum_legend_items} more"
                    )
                )
            )

        ax.legend(
            handles=handles_to_show,
            loc="upper left",
            bbox_to_anchor=(1.02, 1.0),
            fontsize=7
        )

    fig.tight_layout()
    fig.savefig(
        image_path,
        format="png",
        dpi=160,
        bbox_inches="tight"
    )
    plt.close(fig)

    image_path = Path(image_path)

    if not image_path.exists() or image_path.stat().st_size == 0:
        raise RuntimeError(
            "Matplotlib fallback did not create a valid PNG file."
        )

    return str(image_path)


# ============================================================
# A4 PDF REPORT EXPORT
# ============================================================

def get_base_item_name(item_name):
    """
    Removes the generated numeric suffix from names such as:
    'Laptop Box-12' -> 'Laptop Box'
    """
    return re.sub(r"-\d+$", "", item_name)


def group_unpacked_boxes(remaining_boxes):
    """
    Groups failed items by base name, dimensions, and weight.
    """
    grouped = Counter()

    for box in remaining_boxes:
        key = (
            get_base_item_name(box.name),
            box.width,
            box.depth,
            box.height,
            box.weight,
            box.stackable
        )
        grouped[key] += 1

    return grouped


def make_report_figure(result):
    """
    Builds a fixed isometric view for inclusion in the PDF.
    The normal interactive figure remains available in Colab.
    """
    main_box = result["main_box"]

    camera = dict(
        eye=dict(x=1.55, y=1.55, z=1.15),
        center=dict(x=0, y=0, z=-0.1),
        up=dict(x=0, y=0, z=1)
    )

    title = (
        f"Truck {result['usage_number']}: {main_box.name}<br>"
        f"{result.get('cargo_group', 'Cargo')} - "
        f"{'Stacking Enabled' if result.get('allow_stacking', False) else 'Floor Only'} | "
        f"Boxes: {len(result['placements'])} | "
        f"Floor utilization: {result['floor_utilization']:.2f}% | "
        f"Weight utilization: {result['weight_utilization']:.2f}%"
    )

    return visualize_3d_packing(
        main_box.width,
        main_box.depth,
        main_box.height,
        result["placements"],
        title=title,
        show=False,
        camera=camera
    )


def create_packing_report_pdf(
    used_main_boxes_3d,
    remaining_boxes,
    total_item_count,
    packed_count,
    item_gap,
    container_gap,
    center_items,
    output_filename="packing_report.pdf"
):
    """
    Creates a multi-page A4 PDF containing:
    - the overall packing summary,
    - the grouped truck breakdown,
    - the grouped unpacked-item breakdown,
    - one detailed section and static 3D image per used truck,
    - placement tables for each truck.

    Plotly remains interactive in the notebook, while the PDF contains
    static snapshots because PDF does not preserve Plotly interaction.
    """
    try:
        from reportlab.lib import colors
        from reportlab.lib.enums import TA_CENTER
        from reportlab.lib.pagesizes import A4
        from reportlab.lib.styles import getSampleStyleSheet, ParagraphStyle
        from reportlab.lib.units import cm
        from reportlab.platypus import (
            SimpleDocTemplate,
            Paragraph,
            Spacer,
            Image,
            Table,
            TableStyle,
            PageBreak,
            LongTable,
            KeepTogether
        )
    except ImportError as exc:
        raise RuntimeError(
            "ReportLab is not installed. In Google Colab, run:\\n"
            "!pip install -q reportlab kaleido"
        ) from exc

    output_path = Path(output_filename).resolve()
    styles = getSampleStyleSheet()

    title_style = ParagraphStyle(
        "ReportTitle",
        parent=styles["Title"],
        alignment=TA_CENTER,
        fontSize=18,
        leading=22,
        spaceAfter=14
    )

    section_style = ParagraphStyle(
        "SectionHeading",
        parent=styles["Heading2"],
        fontSize=13,
        leading=16,
        spaceBefore=8,
        spaceAfter=8
    )

    small_style = ParagraphStyle(
        "SmallText",
        parent=styles["BodyText"],
        fontSize=8,
        leading=10
    )

    normal_style = styles["BodyText"]

    doc = SimpleDocTemplate(
        str(output_path),
        pagesize=A4,
        rightMargin=1.4 * cm,
        leftMargin=1.4 * cm,
        topMargin=1.4 * cm,
        bottomMargin=1.4 * cm,
        title="3D Packing Optimizer Report",
        author="3D Packing Optimizer"
    )

    story = []

    story.append(Paragraph(
        "Multi-Main-Box 3D Packing Report",
        title_style
    ))
    story.append(Paragraph(
        f"Generated: {datetime.now().strftime('%d %B %Y, %H:%M')}",
        normal_style
    ))
    story.append(Spacer(1, 8))

    truck_counts = Counter(
        (
            result.get("cargo_group", "Cargo"),
            result["main_box"].name,
            "Stacking Enabled" if result.get("allow_stacking", False) else "Floor Only"
        )
        for result in used_main_boxes_3d
    )

    summary_data = [
        ["Summary", "Result"],
        ["Total item boxes entered", str(total_item_count)],
        ["Total item boxes packed", str(packed_count)],
        ["Total item boxes not packed", str(len(remaining_boxes))],
        ["Total main boxes used", str(len(used_main_boxes_3d))],
        ["Container clearance", f"{container_gap:.2f} cm"],
        ["Item-to-item gap", f"{item_gap:.2f} cm"],
        ["Center items", str(center_items)],
        [
            "Packing status",
            (
                "All item boxes were packed."
                if not remaining_boxes
                else "Some item boxes could not be packed."
            )
        ]
    ]

    summary_table = Table(
        summary_data,
        colWidths=[7.0 * cm, 10.0 * cm],
        repeatRows=1
    )
    summary_table.setStyle(TableStyle([
        ("BACKGROUND", (0, 0), (-1, 0), colors.HexColor("#D9EAF7")),
        ("TEXTCOLOR", (0, 0), (-1, 0), colors.black),
        ("FONTNAME", (0, 0), (-1, 0), "Helvetica-Bold"),
        ("FONTNAME", (0, 1), (0, -1), "Helvetica-Bold"),
        ("GRID", (0, 0), (-1, -1), 0.5, colors.grey),
        ("VALIGN", (0, 0), (-1, -1), "TOP"),
        ("ROWBACKGROUNDS", (0, 1), (-1, -1), [
            colors.white,
            colors.HexColor("#F5F7F9")
        ]),
        ("FONTSIZE", (0, 0), (-1, -1), 9),
        ("LEFTPADDING", (0, 0), (-1, -1), 6),
        ("RIGHTPADDING", (0, 0), (-1, -1), 6),
        ("TOPPADDING", (0, 0), (-1, -1), 5),
        ("BOTTOMPADDING", (0, 0), (-1, -1), 5)
    ]))
    story.append(summary_table)
    story.append(Spacer(1, 14))

    story.append(Paragraph("Truck Types Used", section_style))

    if truck_counts:
        truck_data = [["Amount", "Truck type", "Cargo group", "Stacking mode"]]
        for (cargo_group, truck_name, stacking_label), count in truck_counts.items():
            truck_data.append([str(count), truck_name, cargo_group, stacking_label])

        truck_table = Table(
            truck_data,
            colWidths=[2.0 * cm, 7.0 * cm, 4.0 * cm, 4.0 * cm],
            repeatRows=1
        )
        truck_table.setStyle(TableStyle([
            ("BACKGROUND", (0, 0), (-1, 0), colors.HexColor("#E4F1E7")),
            ("FONTNAME", (0, 0), (-1, 0), "Helvetica-Bold"),
            ("GRID", (0, 0), (-1, -1), 0.5, colors.grey),
            ("VALIGN", (0, 0), (-1, -1), "TOP"),
            ("FONTSIZE", (0, 0), (-1, -1), 9),
            ("ALIGN", (0, 1), (0, -1), "CENTER")
        ]))
        story.append(truck_table)
    else:
        story.append(Paragraph(
            "No truck was used because no item could be packed.",
            normal_style
        ))

    if remaining_boxes:
        story.append(Spacer(1, 14))
        story.append(Paragraph(
            "Item Boxes That Could Not Fit",
            section_style
        ))

        unpacked_data = [[
            "Item type",
            "Length × Width × Height",
            "Weight/item",
            "Stacking",
            "Amount"
        ]]

        grouped_unpacked = group_unpacked_boxes(remaining_boxes)

        for (
            name,
            width,
            depth,
            height,
            weight,
            stackable
        ), count in grouped_unpacked.items():
            unpacked_data.append([
                name,
                f"{width:.1f} × {depth:.1f} × {height:.1f} cm",
                f"{weight:.1f} kg",
                "Stackable" if stackable else "Unstackable",
                str(count)
            ])

        unpacked_table = LongTable(
            unpacked_data,
            colWidths=[
                4.5 * cm,
                5.8 * cm,
                2.5 * cm,
                2.6 * cm,
                1.6 * cm
            ],
            repeatRows=1
        )
        unpacked_table.setStyle(TableStyle([
            ("BACKGROUND", (0, 0), (-1, 0), colors.HexColor("#F7DFDF")),
            ("FONTNAME", (0, 0), (-1, 0), "Helvetica-Bold"),
            ("GRID", (0, 0), (-1, -1), 0.5, colors.grey),
            ("VALIGN", (0, 0), (-1, -1), "TOP"),
            ("FONTSIZE", (0, 0), (-1, -1), 8),
            ("ALIGN", (-1, 1), (-1, -1), "CENTER")
        ]))
        story.append(unpacked_table)

    with tempfile.TemporaryDirectory() as temp_dir:
        temp_path = Path(temp_dir)

        for result in used_main_boxes_3d:
            main_box = result["main_box"]
            usage_number = result["usage_number"]

            story.append(PageBreak())
            story.append(Paragraph(
                f"Used Main Box {usage_number}: {main_box.name}",
                section_style
            ))

            detail_data = [
                ["Truck detail", "Value"],
                ["Cargo group", result.get("cargo_group", "Cargo")],
                [
                    "Stacking mode",
                    (
                        "Stacking Enabled"
                        if result.get("allow_stacking", False)
                        else "Floor Only"
                    )
                ],
                [
                    "Cargo dimensions",
                    (
                        f"{main_box.width:.1f} × "
                        f"{main_box.depth:.1f} × "
                        f"{main_box.height:.1f} cm"
                    )
                ],
                ["Maximum weight", f"{main_box.max_weight:.2f} kg"],
                ["Boxes packed", str(len(result["placements"]))],
                ["Used floor area", f"{result['fitted_floor_area']:.2f} cm²"],
                [
                    "Available floor area",
                    f"{main_box.width * main_box.depth:.2f} cm²"
                ],
                [
                    "Floor utilization",
                    f"{result['floor_utilization']:.2f}%"
                ],
                ["Packed weight", f"{result['fitted_weight']:.2f} kg"],
                [
                    "Weight utilization",
                    f"{result['weight_utilization']:.2f}%"
                ],
                ["Used volume", f"{result['fitted_volume']:.2f} cm³"],
                [
                    "Volume utilization",
                    f"{result['volume_utilization']:.2f}%"
                ]
            ]

            detail_table = Table(
                detail_data,
                colWidths=[6.0 * cm, 11.0 * cm],
                repeatRows=1
            )
            detail_table.setStyle(TableStyle([
                ("BACKGROUND", (0, 0), (-1, 0), colors.HexColor("#D9EAF7")),
                ("FONTNAME", (0, 0), (-1, 0), "Helvetica-Bold"),
                ("FONTNAME", (0, 1), (0, -1), "Helvetica-Bold"),
                ("GRID", (0, 0), (-1, -1), 0.5, colors.grey),
                ("VALIGN", (0, 0), (-1, -1), "TOP"),
                ("FONTSIZE", (0, 0), (-1, -1), 8.5)
            ]))
            story.append(detail_table)
            story.append(Spacer(1, 10))

            figure = make_report_figure(result)
            image_path = temp_path / f"truck_{usage_number}.png"

            try:
                export_plotly_figure_to_png(
                    figure=figure,
                    image_path=image_path,
                    width=1300,
                    height=850,
                    scale=1
                )

            except Exception as kaleido_error:
                print(
                    "\nPlotly/Kaleido could not create the PDF image. "
                    "Using the Matplotlib fallback instead."
                )
                print(kaleido_error)

                try:
                    save_matplotlib_truck_image(
                        result=result,
                        image_path=image_path
                    )

                except Exception as fallback_error:
                    raise RuntimeError(
                        "Both static-image export methods failed.\n\n"
                        "PLOTLY/KALEIDO ERROR:\n"
                        f"{kaleido_error}\n\n"
                        "MATPLOTLIB FALLBACK ERROR:\n"
                        f"{fallback_error}\n\n"
                        "Full fallback traceback:\n"
                        f"{traceback.format_exc()}"
                    ) from fallback_error

            report_image = Image(
                str(image_path),
                width=17.0 * cm,
                height=11.1 * cm
            )
            story.append(report_image)
            story.append(Spacer(1, 10))

            story.append(Paragraph("Box Placements", section_style))

            placement_data = [[
                "Item",
                "x",
                "y",
                "z",
                "L",
                "W",
                "H",
                "kg",
                "Stack",
                "Rot."
            ]]

            for placement in result["placements"]:
                placement_data.append([
                    Paragraph(placement.box.name, small_style),
                    f"{placement.x:.1f}",
                    f"{placement.y:.1f}",
                    f"{placement.z:.1f}",
                    f"{placement.width:.1f}",
                    f"{placement.depth:.1f}",
                    f"{placement.height:.1f}",
                    f"{placement.box.weight:.1f}",
                    "Yes" if placement.box.stackable else "No",
                    "Yes" if placement.rotated else "No"
                ])

            placement_table = LongTable(
                placement_data,
                colWidths=[
                    4.1 * cm,
                    1.25 * cm,
                    1.25 * cm,
                    1.25 * cm,
                    1.25 * cm,
                    1.25 * cm,
                    1.25 * cm,
                    1.35 * cm,
                    1.35 * cm,
                    1.15 * cm
                ],
                repeatRows=1
            )
            placement_table.setStyle(TableStyle([
                ("BACKGROUND", (0, 0), (-1, 0), colors.HexColor("#EAEAEA")),
                ("FONTNAME", (0, 0), (-1, 0), "Helvetica-Bold"),
                ("GRID", (0, 0), (-1, -1), 0.35, colors.grey),
                ("VALIGN", (0, 0), (-1, -1), "MIDDLE"),
                ("ALIGN", (1, 1), (-1, -1), "CENTER"),
                ("FONTSIZE", (0, 0), (-1, -1), 7),
                ("TOPPADDING", (0, 0), (-1, -1), 3),
                ("BOTTOMPADDING", (0, 0), (-1, -1), 3)
            ]))
            story.append(placement_table)

        doc.build(story)

    return str(output_path)


def download_file_in_colab(file_path):
    """
    Starts a browser download when running in Google Colab.
    Outside Colab, it prints the saved file location.
    """
    try:
        from google.colab import files
        files.download(file_path)
    except ImportError:
        print(f"PDF saved to: {file_path}")


# ============================================================
# CSV / EXCEL ITEM INPUT
# ============================================================

def _normalize_dataframe_columns(dataframe):
    """Returns a case-insensitive mapping of cleaned column names."""
    return {
        str(column).strip().lower(): column
        for column in dataframe.columns
    }


def _validate_and_expand_item_dataframe(
    dataframe,
    source_name="DataFrame",
    skip_invalid_rows=False
):
    """
    Validates item data and expands Quantity into individual Box3D objects.

    Required columns (case-insensitive):
    - Name
    - Length
    - Width
    - Height
    - Quantity
    - Weight
    - Stackable

    When skip_invalid_rows=True, rows such as repeated PDF headers or blank
    extraction artifacts are skipped and reported instead of stopping the run.
    """
    if not isinstance(dataframe, pd.DataFrame):
        raise TypeError(f"{source_name} must be a pandas DataFrame.")

    if dataframe.empty:
        raise ValueError(f"{source_name} is empty.")

    dataframe = dataframe.copy()
    normalized_columns = _normalize_dataframe_columns(dataframe)

    required_columns = [
        "name", "length", "width", "height",
        "quantity", "weight", "stackable"
    ]

    missing_columns = [
        column for column in required_columns
        if column not in normalized_columns
    ]

    if missing_columns:
        raise ValueError(
            f"{source_name} is missing these required columns: "
            + ", ".join(column.title() for column in missing_columns)
        )

    item_boxes_3d = []
    skipped_rows = []

    for dataframe_index, row in dataframe.iterrows():
        source_row_number = dataframe_index + 1

        if row.isna().all():
            continue

        try:
            name_value = row[normalized_columns["name"]]
            name = "" if pd.isna(name_value) else str(name_value).strip()

            if not name:
                name = f"ItemBoxRow{source_row_number}"

            length = float(row[normalized_columns["length"]])
            width = float(row[normalized_columns["width"]])
            height = float(row[normalized_columns["height"]])
            quantity_value = float(row[normalized_columns["quantity"]])
            weight = float(row[normalized_columns["weight"]])
            stackable = parse_stackable_value(
                row[normalized_columns["stackable"]]
            )

            if not quantity_value.is_integer():
                raise ValueError("Quantity must be a whole number.")

            quantity = int(quantity_value)

            if length <= 0:
                raise ValueError("Length must be greater than zero.")
            if width <= 0:
                raise ValueError("Width must be greater than zero.")
            if height <= 0:
                raise ValueError("Height must be greater than zero.")
            if quantity <= 0:
                raise ValueError("Quantity must be greater than zero.")
            if weight <= 0:
                raise ValueError("Weight must be greater than zero.")

        except (TypeError, ValueError) as error:
            if skip_invalid_rows:
                skipped_rows.append((source_row_number, str(error)))
                continue

            raise ValueError(
                f"Invalid data in {source_name} row "
                f"{source_row_number}: {error}"
            ) from error

        for item_number in range(1, quantity + 1):
            item_boxes_3d.append(
                Box3D(
                    name=f"{name}-{item_number}",
                    width=length,
                    depth=width,
                    height=height,
                    weight=weight,
                    stackable=stackable
                )
            )

    if skipped_rows:
        print(
            f"Skipped {len(skipped_rows)} invalid or non-data row(s) "
            f"from {source_name}."
        )
        preview_limit = 10
        for row_number, reason in skipped_rows[:preview_limit]:
            print(f"  - Row {row_number}: {reason}")
        if len(skipped_rows) > preview_limit:
            print(
                f"  ... and {len(skipped_rows) - preview_limit} more skipped rows."
            )

    if not item_boxes_3d:
        raise ValueError(
            f"{source_name} does not contain any valid item boxes."
        )

    return item_boxes_3d


def load_item_boxes_from_file(file_path):
    """Reads item-box data from a CSV or Excel file."""
    file_path = Path(file_path)

    if not file_path.exists():
        raise FileNotFoundError(f"Input file was not found: {file_path}")

    extension = file_path.suffix.lower()

    if extension == ".csv":
        dataframe = pd.read_csv(file_path)
    elif extension in [".xlsx", ".xls"]:
        dataframe = pd.read_excel(file_path)
    else:
        raise ValueError("Unsupported file type. Use CSV, XLSX, or XLS.")

    return _validate_and_expand_item_dataframe(
        dataframe,
        source_name=f"CSV/Excel file '{file_path.name}'",
        skip_invalid_rows=False
    )


def load_item_boxes_from_dataframe(
    dataframe,
    source_name="Existing Colab DataFrame"
):
    """
    Reads item-box data directly from an existing pandas DataFrame.

    Invalid extraction artifacts, such as repeated PDF header rows, are skipped.
    """
    return _validate_and_expand_item_dataframe(
        dataframe,
        source_name=source_name,
        skip_invalid_rows=True
    )


def _choose_dataframe_column(dataframe, field_name, allow_default=False):
    """Asks the user which DataFrame column represents one optimizer field."""
    columns = list(dataframe.columns)

    while True:
        default_text = " or D for a default value" if allow_default else ""
        answer = input(
            f"Column number for {field_name}{default_text}: "
        ).strip()

        if allow_default and answer.lower() in ["d", "default"]:
            return None

        try:
            column_number = int(answer)
        except ValueError:
            print("Please enter a valid column number.")
            continue

        if 1 <= column_number <= len(columns):
            return columns[column_number - 1]

        print(f"Please enter a number from 1 to {len(columns)}.")


def prepare_existing_dataframe_for_optimizer(dataframe):
    """
    Prepares a Camelot/editable-grid DataFrame for the packing optimizer.

    If the required optimizer column names already exist, they are used directly.
    Otherwise, the user maps the existing columns once through numbered prompts.
    """
    if not isinstance(dataframe, pd.DataFrame):
        raise TypeError("The selected runtime object is not a pandas DataFrame.")

    if dataframe.empty:
        raise ValueError("The selected runtime DataFrame is empty.")

    dataframe = dataframe.copy().reset_index(drop=True)

    # Remove ipydatagrid-generated index columns when present.
    for possible_index_column in ["index", "level_0"]:
        if possible_index_column in dataframe.columns:
            dataframe = dataframe.drop(columns=[possible_index_column])

    required_columns = [
        "Name", "Length", "Width", "Height",
        "Quantity", "Weight", "Stackable"
    ]

    normalized_columns = _normalize_dataframe_columns(dataframe)

    if all(column.lower() in normalized_columns for column in required_columns):
        prepared = dataframe[
            [normalized_columns[column.lower()] for column in required_columns]
        ].copy()
        prepared.columns = required_columns
        print("Required optimizer columns were detected automatically.")
        return prepared

    print("\nThe existing DataFrame does not use the optimizer column names.")
    print("Map the extracted columns to the required optimizer fields.")
    print("\nAvailable columns:")

    for index, column in enumerate(dataframe.columns, start=1):
        sample_values = (
            dataframe[column]
            .dropna()
            .astype(str)
            .str.replace("\n", " ", regex=False)
            .head(3)
            .tolist()
        )
        sample_text = " | ".join(sample_values)
        print(f"  {index}. {column}    Sample: {sample_text[:120]}")

    mapping = {
        "Name": _choose_dataframe_column(dataframe, "Name"),
        "Quantity": _choose_dataframe_column(dataframe, "Quantity"),
        "Length": _choose_dataframe_column(dataframe, "Length"),
        "Width": _choose_dataframe_column(dataframe, "Width"),
        "Height": _choose_dataframe_column(dataframe, "Height"),
        "Weight": _choose_dataframe_column(dataframe, "Weight"),
        "Stackable": _choose_dataframe_column(
            dataframe,
            "Stackable",
            allow_default=True
        )
    }

    prepared = pd.DataFrame()

    for target_column in [
        "Name", "Quantity", "Length", "Width", "Height", "Weight"
    ]:
        prepared[target_column] = dataframe[mapping[target_column]]

    if mapping["Stackable"] is None:
        default_stackable = get_yes_no(
            "Apply Stackable=Yes to every row? yes/no "
            "(No means every row is Unstackable): "
        )
        prepared["Stackable"] = "Yes" if default_stackable else "No"
    else:
        prepared["Stackable"] = dataframe[mapping["Stackable"]]

    print("\nPrepared optimizer DataFrame preview:")
    try:
        from IPython.display import display
        display(prepared.head(20))
    except ImportError:
        print(prepared.head(20))

    return prepared


def get_existing_colab_dataframe():
    """
    Finds table data left in the current Colab runtime by the Camelot program.

    Priority:
    1. editable_grid.data (latest manually edited values)
    2. prepared_dataframe
    3. combined_dataframe
    """
    available_sources = []

    if "editable_grid" in globals():
        try:
            grid_dataframe = editable_grid.data.copy()
            if isinstance(grid_dataframe, pd.DataFrame):
                available_sources.append(
                    ("editable_grid.data (latest edited table)", grid_dataframe)
                )
        except Exception:
            pass

    if "prepared_dataframe" in globals():
        candidate = globals()["prepared_dataframe"]
        if isinstance(candidate, pd.DataFrame):
            available_sources.append(("prepared_dataframe", candidate.copy()))

    if "combined_dataframe" in globals():
        candidate = globals()["combined_dataframe"]
        if isinstance(candidate, pd.DataFrame):
            available_sources.append(("combined_dataframe", candidate.copy()))

    if not available_sources:
        raise NameError(
            "No existing Camelot DataFrame was found in this Colab runtime. "
            "Run the Camelot extraction cell first without restarting the runtime."
        )

    # Remove duplicate source objects by name while preserving priority.
    print("\nExisting runtime table sources:")
    for index, (source_name, dataframe) in enumerate(available_sources, start=1):
        print(
            f"  {index}. {source_name}: "
            f"{len(dataframe)} rows × {len(dataframe.columns)} columns"
        )

    if len(available_sources) == 1:
        selected_name, selected_dataframe = available_sources[0]
    else:
        while True:
            answer = input(
                "Choose the runtime table source number "
                "(press Enter for option 1): "
            ).strip()

            if answer == "":
                selected_index = 0
                break

            try:
                selected_index = int(answer) - 1
            except ValueError:
                print("Please enter a valid source number.")
                continue

            if 0 <= selected_index < len(available_sources):
                break

            print(
                f"Please enter a number from 1 to {len(available_sources)}."
            )

        selected_name, selected_dataframe = available_sources[selected_index]

    print(f"Using runtime source: {selected_name}")
    return selected_dataframe, selected_name


# ============================================================
# INPUT VALIDATION HELPERS (Copied from 2D version)
# ============================================================

def get_positive_float(prompt):
    """
    Repeatedly asks for a positive number.
    """

    while True:
        try:
            value = float(input(prompt))

            if value <= 0:
                print("Please enter a number greater than zero.")
                continue

            return value

        except ValueError:
            print("Invalid input. Please enter a number.")

def get_positive_or_zero_float(prompt):
    """
    Repeatedly asks for a number greater than or equal to zero.
    A value of 0 means no gap.
    """

    while True:
        try:
            value = float(input(prompt))

            if value < 0:
                print("Please enter zero or a positive number.")
                continue

            return value

        except ValueError:
            print("Invalid input. Please enter a number.")


def get_positive_integer(prompt):
    """
    Repeatedly asks for a positive whole number.
    """

    while True:
        try:
            value = int(input(prompt))

            if value <= 0:
                print("Please enter a whole number greater than zero.")
                continue

            return value

        except ValueError:
            print("Invalid input. Please enter a whole number.")


def get_yes_no(prompt):
    """
    Repeatedly asks until the user enters yes or no.
    """

    while True:
        answer = input(prompt).strip().lower()

        if answer in ["yes", "y"]:
            return True

        if answer in ["no", "n"]:
            return False

        print("Please enter yes or no.")


# ============================================================
# MAIN PROGRAM (3D)
# ============================================================

def main_3d():
    print("=== Multi-Main-Box 3D Packing Optimizer ===")

    # --------------------------------------------------------
    # INPUT MAIN-BOX TYPES
    # --------------------------------------------------------

    main_boxes_3d = []

    print("\n=== Using Default Main Box Types from CSV (Unscaled values for display) ===")
    # Predefined Indonesian truck specifications in centimeters
    predefined_truck_specs = [
        {"name": "Pick Up (Bak/Box)", "width": 230.0, "depth": 140.0, "height": 130.0, "max_weight": 1000},
        {"name": "Colt Diesel Engkel (CDE)", "width": 310.0, "depth": 170.0, "height": 170.0, "max_weight": 2000},
        {"name": "Colt Diesel Double (CDD)", "width": 450.0, "depth": 200.0, "height": 200.0, "max_weight": 5000},
        {"name": "CDD Long", "width": 550.0, "depth": 220.0, "height": 220.0, "max_weight": 5000},
        {"name": "Truk Fuso (Engkel Medium)", "width": 600.0, "depth": 250.0, "height": 280.0, "max_weight": 8000},
        {"name": "Truk Tronton", "width": 900.0, "depth": 250.0, "height": 280.0, "max_weight": 15000},
        {"name": "Tronton Wingbox", "width": 900.0, "depth": 250.0, "height": 250.0, "max_weight": 15000},
        {"name": "Trailer Truck 20 Feet", "width": 600.0, "depth": 250.0, "height": 280.0, "max_weight": 25000},
        {"name": "Trailer Truck 40 Feet", "width": 1200.0, "depth": 250.0, "height": 280.0, "max_weight": 25000},
        {"name": "Trailer Lowbet", "width": 700.0, "depth": 300.0, "height": 350.0, "max_weight": 35000},
    ]

    print("All available truck types will be used:")
    for i, truck in enumerate(predefined_truck_specs):
        print(f"  - {truck['name']}: {truck['width']}x{truck['depth']}x{truck['height']} cm, Max Weight: {truck['max_weight']} kg") # Removed score from print

    # Automatically add all predefined truck types
    for truck_data in predefined_truck_specs:
        main_boxes_3d.append(
            MainBox3D(
                name=truck_data["name"],
                width=truck_data["width"],
                depth=truck_data["depth"],
                height=truck_data["height"],
                # score=truck_data["score"],
                max_weight=truck_data["max_weight"]
            )
        )
    print(f"Added {len(main_boxes_3d)} truck types for consideration.")

    # --------------------------------------------------------
    # INPUT ITEM-BOX TYPES
    # --------------------------------------------------------

    item_boxes_3d = []

    print("\n=== Item Box Input Method ===")
    print("1. Enter item boxes manually")
    print("2. Upload a CSV or Excel file")
    print("3. Use a DataFrame already stored in this Colab runtime")

    while True:
        input_method = input(
            "Choose input method (1, 2, or 3): "
        ).strip()

        if input_method in ["1", "2", "3"]:
            break

        print("Please enter 1, 2, or 3.")

    # --------------------------------------------------------
    # OPTION 1: MANUAL INPUT
    # --------------------------------------------------------

    if input_method == "1":
        item_box_type_number = 1

        print("\n=== Enter Item 3D Boxes Manually ===")

        while True:
            print(f"\nItem Box Type {item_box_type_number}")

            name = input(
                "Enter item box name (e.g., 'Laptop Box'): "
            ).strip()

            if not name:
                name = f"ItemBoxType{item_box_type_number}"

            length = get_positive_float(
                "Enter item box length (cm): "
            )
            width = get_positive_float(
                "Enter item box width (cm): "
            )
            height = get_positive_float(
                "Enter item box height (cm): "
            )
            quantity = get_positive_integer(
                "Enter item box quantity: "
            )
            weight = get_positive_float(
                "Enter weight of each item box (kg): "
            )

            stackable = get_yes_no(
                "Is this item type stackable? yes/no: "
            )

            for item_number in range(1, quantity + 1):
                item_boxes_3d.append(
                    Box3D(
                        name=f"{name}-{item_number}",
                        width=length,
                        depth=width,
                        height=height,
                        weight=weight,
                        stackable=stackable
                    )
                )

            add_more = get_yes_no(
                "\nAdd another item box type? yes/no: "
            )

            if not add_more:
                break

            item_box_type_number += 1

    # --------------------------------------------------------
    # OPTION 2: CSV OR EXCEL UPLOAD
    # --------------------------------------------------------

    elif input_method == "2":
        print("\n=== Upload CSV or Excel Item File ===")
        print(
            "Required columns: Name, Length, Width, Height, "
            "Quantity, Weight, Stackable"
        )
        print(
            "Dimensions must be in cm and Weight must be kg per item."
        )

        try:
            from google.colab import files

            uploaded_files = files.upload()

            if not uploaded_files:
                raise ValueError("No file was uploaded.")

            uploaded_filename = next(iter(uploaded_files))
            item_boxes_3d = load_item_boxes_from_file(uploaded_filename)

        except ImportError:
            file_path = input(
                "Enter the CSV or Excel file path: "
            ).strip()
            item_boxes_3d = load_item_boxes_from_file(file_path)

        print(
            f"Successfully loaded {len(item_boxes_3d)} "
            "individual item boxes."
        )

    # --------------------------------------------------------
    # OPTION 3: EXISTING COLAB DATAFRAME
    # --------------------------------------------------------

    else:
        print("\n=== Use Existing Colab DataFrame ===")

        runtime_dataframe, runtime_source_name = (
            get_existing_colab_dataframe()
        )

        prepared_runtime_dataframe = (
            prepare_existing_dataframe_for_optimizer(
                runtime_dataframe
            )
        )

        # Keep the prepared table available for later Colab cells.
        globals()["prepared_dataframe"] = (
            prepared_runtime_dataframe.copy()
        )

        item_boxes_3d = load_item_boxes_from_dataframe(
            prepared_runtime_dataframe,
            source_name=runtime_source_name
        )

        print(
            f"Successfully loaded {len(item_boxes_3d)} "
            "individual item boxes from the existing runtime DataFrame."
        )

    allow_rotation = get_yes_no(
        "\nAllow item box rotation? yes/no: "
    )

    container_gap = get_positive_or_zero_float(
        "\nEnter required gap between item boxes and main container (cm): "
    )

    item_gap = get_positive_or_zero_float(
        "\nEnter required gap between item boxes (cm): "
    )

    center_items = get_yes_no(
        "\nCenter all packed items in the main container? yes/no: "
    )

    # --------------------------------------------------------
    # RUN MULTIPLE MAIN-BOX PACKING (3D)
    # --------------------------------------------------------

    used_main_boxes_3d, remaining_boxes, total_score = (
        pack_separated_stackable_and_unstackable_items(
            main_boxes_3d=main_boxes_3d,
            item_boxes_3d=item_boxes_3d,
            allow_rotation=allow_rotation,
            item_gap=item_gap,
            container_gap=container_gap,
            center_items=center_items
        )
    )

    # --------------------------------------------------------
    # PRINT OVERALL RESULT
    # --------------------------------------------------------

    packed_count = sum(
        len(result["placements"])
        for result in used_main_boxes_3d
    )

    print("\n" + "=" * 55)
    print("MULTI-MAIN-BOX 3D PACKING RESULT")
    print("=" * 55)

    stackable_count = sum(1 for box in item_boxes_3d if box.stackable)
    unstackable_count = len(item_boxes_3d) - stackable_count

    print(f"Total item boxes entered: {len(item_boxes_3d)}")
    print(f"Stackable item boxes entered: {stackable_count}")
    print(f"Unstackable item boxes entered: {unstackable_count}")
    print(f"Total item boxes packed: {packed_count}")
    print(f"Total item boxes not packed: {len(remaining_boxes)}")
    print(f"Total main boxes used: {len(used_main_boxes_3d)}")

    # Add breakdown of used main boxes
    if used_main_boxes_3d:
        truck_counts = Counter(
            (
                result.get("cargo_group", "Cargo"),
                result["main_box"].name,
                "Stacking Enabled" if result.get("allow_stacking", False) else "Floor Only"
            )
            for result in used_main_boxes_3d
        )
        print("\nMain boxes used:")
        for (cargo_group, truck_name, stacking_label), count in truck_counts.items():
            print(f"- {count} × {truck_name} [{cargo_group} - {stacking_label}]")

    print(f"\nRequired gap between item boxes and main container: {container_gap} cm")
    print(f"Required gap between item boxes: {item_gap} cm")
    print(f"Center items: {center_items}")
    # print(f"Total score: {total_score:.2f}") # Removed total score print

    if not remaining_boxes:
        print("Packing status: All item boxes were packed.")
    else:
        print("Packing status: Some item boxes could not be packed.")

    # --------------------------------------------------------
    # PRINT EACH USED MAIN BOX
    # --------------------------------------------------------

    for result in used_main_boxes_3d:
        main_box = result["main_box"]
        placements = result["placements"]

        print("\n" + "-" * 55)
        print(f"USED MAIN BOX {result['usage_number']}")
        print("-" * 55)

        print(f"Cargo group: {result.get('cargo_group', 'Cargo')}")
        print(
            "Stacking mode: "
            + ("Stacking Enabled" if result.get("allow_stacking", False) else "Floor Only")
        )
        print(f"Type: {main_box.name}")
        print(f"Size: {main_box.width} x {main_box.depth} x {main_box.height} cm") # Added depth
        # print(f"Score: {main_box.score}") # Removed score print
        print(f"Maximum weight: {main_box.max_weight} kg")
        print(f"Boxes packed here: {len(placements)}")
        print(f"Used volume: {result['fitted_volume']:.2f} cm³") # Changed to volume
        print(f"Main-box volume: {main_box.volume():.2f} cm³") # Changed to volume
        print(
            f"Used floor area: "
            f"{result['fitted_floor_area']:.2f} cm²"
        )

        print(
            f"Main-box floor area: "
            f"{main_box.width * main_box.depth:.2f} cm²"
        )

        print(
            f"Floor utilization: "
            f"{result['floor_utilization']:.2f}%"
        )

        print(
            f"Volume utilization: "
            f"{result['volume_utilization']:.2f}%"
        )
        print(f"Packed weight: {result['fitted_weight']:.2f} kg")
        print(f"Weight utilization: {result['weight_utilization']:.2f} %\n")

        print("Box placements:")

        for placement in placements:
            print(
                f"{placement.box.name}: "
                f"x={placement.x:.2f} cm, "
                f"y={placement.y:.2f} cm, "
                f"z={placement.z:.2f} cm, " # Added z
                f"w={placement.width:.2f} cm, "
                f"d={placement.depth:.2f} cm, " # Added depth
                f"h={placement.height:.2f} cm, "
                f"weight={placement.box.weight:.2f} kg, " # Added weight
                f"stackable={placement.box.stackable}, "
                f"rotated={placement.rotated}"
            )

        # Removed print_3d_layers call as requested by the user.
        # print_3d_layers(
        #     main_box.width,
        #     main_box.depth,
        #     main_box.height,
        #     placements,
        #     main_box_name=(
        #         f"Main Box {result['usage_number']} "
        #         f"- {main_box.name}"
        #     )
        # )

    # --------------------------------------------------------
    # PRINT ITEMS THAT COULD NOT FIT ANY MAIN BOX
    # --------------------------------------------------------

    if remaining_boxes:
        print("\n" + "=" * 55)
        print("Item Boxes That Could Not Fit")
        print("=" * 55)

        # Group remaining boxes by type, dimensions, and weight
        grouped_unpacked = Counter()
        for box in remaining_boxes:
            # Remove numbering suffix (e.g., "-1", "-2")
            base_name = re.sub(r'-\d+$', '', box.name)
            key = (
                base_name,
                box.width,
                box.depth,
                box.height,
                box.weight,
                box.stackable
            )
            grouped_unpacked[key] += 1

        for (
            name,
            width,
            depth,
            height,
            weight,
            stackable
        ), count in grouped_unpacked.items():
            stack_label = "Stackable" if stackable else "Unstackable"
            print(
                f"- {name}: "
                f"{width:.1f} × {depth:.1f} × {height:.1f} cm, "
                f"Weight: {weight:.1f} kg, "
                f"Stacking: {stack_label}, "
                f"Amount: {count}"
            )

    # --------------------------------------------------------
    # DISPLAY ONE PLOTLY FIGURE PER USED MAIN BOX
    # --------------------------------------------------------

    visualize_multiple_main_boxes_3d(used_main_boxes_3d)

    # --------------------------------------------------------
    # OPTIONAL A4 PDF EXPORT
    # --------------------------------------------------------

    export_pdf = get_yes_no(
        "\nExport the complete result to an A4 PDF? yes/no: "
    )

    if export_pdf:
        try:
            pdf_path = create_packing_report_pdf(
                used_main_boxes_3d=used_main_boxes_3d,
                remaining_boxes=remaining_boxes,
                total_item_count=len(item_boxes_3d),
                packed_count=packed_count,
                item_gap=item_gap,
                container_gap=container_gap,
                center_items=center_items,
                output_filename="packing_report.pdf"
            )

            print(f"PDF report created successfully: {pdf_path}")
            print("The PDF contains static snapshots; the notebook plots remain interactive.")
            download_file_in_colab(pdf_path)

        except Exception as error:
            print("\nPDF export failed.")
            print(error)

    # --------------------------------------------------------
    # OPTIONAL INTERACTIVE HTML EXPORT
    # --------------------------------------------------------

    export_html = get_yes_no(
        "\nExport the complete result to an interactive HTML report? yes/no: "
    )

    if export_html:
        try:
            html_report_path = create_packing_report_html(
                used_main_boxes_3d=used_main_boxes_3d,
                remaining_boxes=remaining_boxes,
                total_item_count=len(item_boxes_3d),
                packed_count=packed_count,
                item_gap=item_gap,
                container_gap=container_gap,
                center_items=center_items,
                output_filename="packing_report.html"
            )
            print(f"Interactive HTML report created successfully: {html_report_path}")
            print("Open it in a browser to rotate, zoom, pan, and inspect the 3D views.")
            download_file_in_colab(html_report_path)
        except Exception as error:
            print("\nHTML export failed.")
            print(error)


# Run the program.
main_3d()
```

```python
# 3D Packing Optimizer Version 1.8 — Test separate stackable/unstackable trucks with stacking mode alongside HTML export (Master Code)
import plotly.graph_objects as go
import plotly.io as pio
from collections import Counter
from datetime import datetime
from pathlib import Path
import os
import re
import shutil
import tempfile
import traceback
import pandas as pd

pio.renderers.default = "colab"


# ============================================================
# DATA CLASSES
# ============================================================

class Box3D:
    """
    Represents an item box with dimensions, weight, and stacking permission.

    stackable:
    - False / unstackable: must be loaded into floor-only trucks.
    - True / stackable: loaded separately into stackable-only trucks, where
      vertical stacking is allowed.
    """

    def __init__(self, name, width, depth, height, weight, stackable=False):
        self.name = name
        self.width = width
        self.depth = depth
        self.height = height
        self.weight = weight
        self.stackable = stackable

    def volume(self):
        return self.width * self.depth * self.height


class Placement3D:
    """
    Stores the final position and orientation of a packed item box.
    """

    def __init__(self, box, x, y, z, width, depth, height, rotated=False):
        self.box = box
        self.x = x
        self.y = y
        self.z = z
        self.width = width
        self.depth = depth
        self.height = height
        self.rotated = rotated


class MainBox3D:
    """
    Represents a reusable main 3D container with:
    - dimensions,
    - score/cost,
    - maximum weight capacity.
    """

    def __init__(self, name, width, depth, height, max_weight):
        self.name = name
        self.width = width
        self.depth = depth
        self.height = height
        self.max_weight = max_weight

    def volume(self):
        return self.width * self.depth * self.height


# ============================================================
# BASIC PACKING HELPERS
# ============================================================

def volume(space):
    """
    Calculates the volume of a free-space in 3D.
    """
    return space["w"] * space["d"] * space["h"]


def get_orientations(box, allow_rotation=True):
    """
    Returns possible orientations of a 3D box.
    Width, depth, and height can be rearranged if rotation is allowed.
    """
    if not allow_rotation:
        # Original dimensions (width, depth, height, not rotated)
        return [(box.width, box.depth, box.height, False)]

    dims = [
        (box.width, box.depth, box.height),    # Original
        (box.width, box.height, box.depth),    # W H D
        (box.depth, box.width, box.height),    # D W H
        (box.depth, box.height, box.width),    # D H W
        (box.height, box.width, box.depth),    # H W D
        (box.height, box.depth, box.width),    # H D W
    ]

    unique_dims = []
    for dim in dims:
        if dim not in unique_dims:
            unique_dims.append(dim)

    orientations = []
    for w, d, h in unique_dims:
        # A box is considered "rotated" if its orientation is different from its original definition.
        rotated = (w, d, h) != (box.width, box.depth, box.height)
        orientations.append((w, d, h, rotated))

    return orientations


def can_fit_3d(width, depth, height, space):
    """
    Checks whether a box of given dimensions fits in a 3D space.
    """
    return (
        width <= space["w"] and
        depth <= space["d"] and
        height <= space["h"]
    )


def parse_stackable_value(value):
    """
    Converts manual/CSV stackable values into True/False.

    Accepted stackable values:
    - yes, y, true, t, 1, stackable

    Accepted unstackable values:
    - no, n, false, f, 0, unstackable, non-stackable, not stackable
    """
    if isinstance(value, bool):
        return value

    value_text = str(value).strip().lower()

    stackable_values = {
        "yes", "y", "true", "t", "1", "stackable"
    }

    unstackable_values = {
        "no", "n", "false", "f", "0",
        "unstackable", "non-stackable", "non stackable",
        "not stackable"
    }

    if value_text in stackable_values:
        return True

    if value_text in unstackable_values:
        return False

    raise ValueError(
        "Stackable must be yes/no, true/false, 1/0, "
        "stackable, or unstackable."
    )


def split_space_3d_floor_only(
    space,
    used_width,
    used_depth,
    item_gap
):
    """
    Splits the remaining floor area after placing an item.

    No space is created above the item.
    Every free space remains at z = 0.

    The split covers the entire remaining floor without overlap.
    """

    new_spaces = []

    occupied_width = min(
        used_width + item_gap,
        space["w"]
    )

    occupied_depth = min(
        used_depth + item_gap,
        space["d"]
    )

    # Right space uses the full remaining depth.
    right_space = {
        "x": space["x"] + occupied_width,
        "y": space["y"],
        "z": 0.0,
        "w": space["w"] - occupied_width,
        "d": space["d"],
        "h": space["h"]
    }

    # Front space only uses the width occupied by the item.
    # This prevents overlap with right_space.
    front_space = {
        "x": space["x"],
        "y": space["y"] + occupied_depth,
        "z": 0.0,
        "w": occupied_width,
        "d": space["d"] - occupied_depth,
        "h": space["h"]
    }

    for new_space in [right_space, front_space]:
        if (
            new_space["w"] > 0
            and new_space["d"] > 0
            and new_space["h"] > 0
        ):
            new_spaces.append(new_space)

    return new_spaces


def split_space_3d_with_stacking(
    space,
    used_width,
    used_depth,
    used_height,
    item_gap
):
    """
    Splits remaining 3D space after placing an item.

    This version is only used for stackable-only trucks.

    It creates:
    - right space,
    - front space,
    - upper space.

    The upper space allows another stackable item to be placed above the
    current stackable item.
    """

    new_spaces = []

    occupied_width = min(
        used_width + item_gap,
        space["w"]
    )

    occupied_depth = min(
        used_depth + item_gap,
        space["d"]
    )

    occupied_height = min(
        used_height + item_gap,
        space["h"]
    )

    right_space = {
        "x": space["x"] + occupied_width,
        "y": space["y"],
        "z": space["z"],
        "w": space["w"] - occupied_width,
        "d": space["d"],
        "h": space["h"]
    }

    front_space = {
        "x": space["x"],
        "y": space["y"] + occupied_depth,
        "z": space["z"],
        "w": occupied_width,
        "d": space["d"] - occupied_depth,
        "h": space["h"]
    }

    upper_space = {
        "x": space["x"],
        "y": space["y"],
        "z": space["z"] + occupied_height,
        "w": used_width,
        "d": used_depth,
        "h": space["h"] - occupied_height
    }

    for new_space in [right_space, front_space, upper_space]:
        if (
            new_space["w"] > 0
            and new_space["d"] > 0
            and new_space["h"] > 0
        ):
            new_spaces.append(new_space)

    return new_spaces


# ============================================================
# SINGLE MAIN-BOX PACKING ALGORITHM (3D)
# ============================================================

def pack_boxes_3d(
    container_width,
    container_depth,
    container_height,
    boxes,
    max_weight,
    allow_rotation=True,
    item_gap=0.0,
    container_gap=0.0,
    allow_stacking=False
):
    """
    Packs item boxes into one truck.

    If allow_stacking is False:
    - strict floor loading is used,
    - every item is placed at z = 0,
    - no item may be placed above another item.

    If allow_stacking is True:
    - upper spaces may be created,
    - items may be placed at z > 0,
    - this should only be used for stackable-only item groups.

    Weight, rotation, item gaps, and container gaps remain active.
    """

    sorted_boxes = boxes.copy()

    # Larger floor footprints are attempted first.
    sorted_boxes.sort(
        key=lambda box: (
            box.width * box.depth,
            box.volume()
        ),
        reverse=True
    )

    # Container gap applies to the side walls and the ceiling.
    # The floor itself stays at z = 0.
    effective_width = container_width - (2 * container_gap)
    effective_depth = container_depth - (2 * container_gap)
    effective_height = container_height - container_gap

    if (
        effective_width <= 0
        or effective_depth <= 0
        or effective_height <= 0
    ):
        return [], boxes.copy(), 0.0

    # Every free space represents an available floor region.
    free_spaces = [
        {
            "x": container_gap,
            "y": container_gap,
            "z": 0.0,
            "w": effective_width,
            "d": effective_depth,
            "h": effective_height
        }
    ]

    placements = []
    unpacked = []
    current_weight = 0.0

    for box in sorted_boxes:
        best_space_index = None
        best_orientation = None
        best_waste = None

        # Weight-capacity constraint.
        if current_weight + box.weight > max_weight:
            unpacked.append(box)
            continue

        for index, space in enumerate(free_spaces):

            # In floor-only mode, free spaces must stay at z = 0.
            # In stacking mode, z > 0 spaces are allowed.
            if not allow_stacking and space["z"] != 0:
                continue

            for width, depth, height, rotated in get_orientations(
                box,
                allow_rotation
            ):
                # The item must fit in floor width/depth and truck height.
                if can_fit_3d(
                    width,
                    depth,
                    height,
                    space
                ):
                    # Floor-only mode compares unused floor area.
                    # Stacking mode compares unused volume so upper spaces can
                    # be used more naturally.
                    if allow_stacking:
                        waste = (
                            space["w"] * space["d"] * space["h"]
                        ) - (
                            width * depth * height
                        )
                    else:
                        waste = (
                            space["w"] * space["d"]
                        ) - (
                            width * depth
                        )

                    if (
                        best_waste is None
                        or waste < best_waste
                    ):
                        best_waste = waste
                        best_space_index = index
                        best_orientation = (
                            width,
                            depth,
                            height,
                            rotated
                        )

        # No suitable floor region exists in this truck.
        if best_space_index is None:
            unpacked.append(box)
            continue

        selected_space = free_spaces.pop(best_space_index)

        used_width, used_depth, used_height, rotated = (
            best_orientation
        )

        placement = Placement3D(
            box=box,
            x=selected_space["x"],
            y=selected_space["y"],

            # Floor-only items stay at z=0. Stackable-only trucks may use z>0.
            z=selected_space["z"] if allow_stacking else 0.0,

            width=used_width,
            depth=used_depth,
            height=used_height,
            rotated=rotated
        )

        placements.append(placement)
        current_weight += box.weight

        if allow_stacking:
            # Stackable-only trucks may create right, front, and upper spaces.
            new_spaces = split_space_3d_with_stacking(
                selected_space,
                used_width,
                used_depth,
                used_height,
                item_gap
            )
        else:
            # Unstackable trucks only create floor spaces.
            new_spaces = split_space_3d_floor_only(
                selected_space,
                used_width,
                used_depth,
                item_gap
            )

        free_spaces.extend(new_spaces)

        if allow_stacking:
            # Prefer lower spaces first, then smaller volumes.
            free_spaces.sort(
                key=lambda space: (
                    space["z"],
                    space["w"] * space["d"] * space["h"]
                )
            )
        else:
            # Try smaller floor spaces first.
            free_spaces.sort(
                key=lambda space: space["w"] * space["d"]
            )

    return placements, unpacked, current_weight

def center_placements_3d(placements, container_width, container_depth, container_height, container_gap=0.0):
    """
    Centers a group of placements within the available space of a container.
    The available space considers the container_gap.
    This function modifies the placements in place.
    """
    if not placements:
        return placements

    # Calculate the bounding box of all placed items
    min_x = min(p.x for p in placements)
    min_y = min(p.y for p in placements)

    max_x = max(p.x + p.width for p in placements)
    max_y = max(p.y + p.depth for p in placements)

    group_width = max_x - min_x
    group_depth = max_y - min_y

    # Calculate effective container dimensions for centering
    effective_width = container_width - (2 * container_gap)
    effective_depth = container_depth - (2 * container_gap)

    # Calculate offsets to center the group
    # The new 'start' for the group should be container_gap + (effective_dim - group_dim) / 2
    # The offset to apply to each item's coordinate is (new_start - current_min)
    offset_x = container_gap + (effective_width - group_width) / 2 - min_x
    offset_y = container_gap + (effective_depth - group_depth) / 2 - min_y

    for p in placements:
        p.x += offset_x
        p.y += offset_y

    return placements

# ============================================================
# MULTIPLE MAIN-BOX SELECTION ALGORITHM (3D)
# ============================================================

def pack_multiple_main_boxes_3d(
    main_boxes_3d,
    item_boxes_3d,
    allow_rotation=True,
    item_gap=0.0,
    container_gap=0.0,
    center_items=False,
    allow_stacking=False,
    cargo_group_name="Cargo"
):
    """
    Repeatedly chooses and uses 3D main boxes until all possible item boxes
    have been packed.

    Main-box types may be used repeatedly.

    Selection logic:
    1. Try every available main-box type on the remaining item boxes.
    2. If a truck can pack all remaining items, choose the first such truck
       based on the truck list order.
    3. Otherwise, choose the truck that packs the most volume.
    4. Remove the packed item boxes.
    5. Repeat until all item boxes are packed or no main box can
       fit any remaining item.

    This is a greedy heuristic.
    """

    remaining_boxes = item_boxes_3d.copy()
    used_main_boxes = []
    # total_score = 0.0 # Score is no longer used
    usage_number = 1

    while remaining_boxes:
        candidates = []

        for main_box in main_boxes_3d:
            test_placements, test_unpacked, fitted_weight = pack_boxes_3d(
                main_box.width,
                main_box.depth,
                main_box.height,
                remaining_boxes,
                main_box.max_weight,
                allow_rotation,
                item_gap,
                container_gap,
                allow_stacking
            )

            fitted_count = len(test_placements)

            if fitted_count == 0:
                continue

            # Apply centering if requested
            if center_items:
                test_placements = center_placements_3d(
                    test_placements,
                    main_box.width,
                    main_box.depth,
                    main_box.height,
                    container_gap
                )

            fitted_floor_area = sum(
                placement.width * placement.depth
                for placement in test_placements
            )

            main_box_floor_area = (
                main_box.width * main_box.depth
            )

            floor_utilization = (
                fitted_floor_area / main_box_floor_area
            ) * 100

            fitted_volume = sum(
                p.width * p.depth * p.height
                for p in test_placements
            )

            volume_utilization = (fitted_volume / main_box.volume()) * 100
            weight_utilization = (fitted_weight / main_box.max_weight) * 100

            # Efficiency: packed item volume per score point (score is removed, so this is no longer calculated for selection)
            # efficiency = fitted_volume / main_box.score

            candidates.append({
                "main_box": main_box,
                "placements": test_placements,
                "unpacked": test_unpacked,
                "fitted_count": fitted_count,
                "fitted_volume": fitted_volume,
                "fitted_floor_area": fitted_floor_area, # Added fitted_floor_area
                "fitted_weight": fitted_weight,
                "floor_utilization": floor_utilization, # Added floor_utilization
                "volume_utilization": volume_utilization, # Renamed from 'utilization'
                "weight_utilization": weight_utilization
                # "efficiency": efficiency # Removed as score is no longer used for selection
            })

        if not candidates:
            break

        # Order-based logic: Prioritize the first truck that can pack everything
        complete_candidates = [c for c in candidates if len(c["unpacked"]) == 0]

        if complete_candidates:
            # Select the first complete candidate found based on the main_boxes order (CSV order)
            best_choice = complete_candidates[0]
        else:
            # Fallback to the one that fits the most volume, then fitted count as tie-breaker
            best_choice = max(
                candidates,
                key=lambda candidate: (
                    candidate["fitted_volume"],
                    candidate["fitted_count"]
                )
            )

        selected_main_box = best_choice["main_box"]

        used_main_boxes.append({
            "usage_number": usage_number,
            "cargo_group": cargo_group_name,
            "allow_stacking": allow_stacking,
            "main_box": selected_main_box,
            "placements": best_choice["placements"],
            "fitted_floor_area": best_choice["fitted_floor_area"],
            "fitted_volume": best_choice["fitted_volume"],
            "fitted_weight": best_choice["fitted_weight"],
            "floor_utilization": best_choice["floor_utilization"],
            "volume_utilization": best_choice["volume_utilization"],
            "weight_utilization": best_choice["weight_utilization"]
            # "score": selected_main_box.score # Score is no longer used
        })

        # total_score += selected_main_box.score # Score is no longer used
        remaining_boxes = best_choice["unpacked"]
        usage_number += 1

    # Removed total_score calculation since score is not used
    return used_main_boxes, remaining_boxes, 0.0 # Return 0.0 for total_score


def pack_separated_stackable_and_unstackable_items(
    main_boxes_3d,
    item_boxes_3d,
    allow_rotation=True,
    item_gap=0.0,
    container_gap=0.0,
    center_items=False
):
    """
    Separates items into two independent truck groups.

    1. Unstackable items:
       - loaded into their own trucks,
       - floor-only packing,
       - z always stays 0.

    2. Stackable items:
       - loaded into their own trucks,
       - stacking-enabled packing,
       - z may be greater than 0.

    The two groups are never mixed inside the same truck.
    """
    unstackable_boxes = [
        box for box in item_boxes_3d
        if not box.stackable
    ]

    stackable_boxes = [
        box for box in item_boxes_3d
        if box.stackable
    ]

    used_unstackable = []
    remaining_unstackable = []
    used_stackable = []
    remaining_stackable = []

    if unstackable_boxes:
        used_unstackable, remaining_unstackable, _ = pack_multiple_main_boxes_3d(
            main_boxes_3d=main_boxes_3d,
            item_boxes_3d=unstackable_boxes,
            allow_rotation=allow_rotation,
            item_gap=item_gap,
            container_gap=container_gap,
            center_items=center_items,
            allow_stacking=False,
            cargo_group_name="Unstackable"
        )

    if stackable_boxes:
        used_stackable, remaining_stackable, _ = pack_multiple_main_boxes_3d(
            main_boxes_3d=main_boxes_3d,
            item_boxes_3d=stackable_boxes,
            allow_rotation=allow_rotation,
            item_gap=item_gap,
            container_gap=container_gap,
            center_items=center_items,
            allow_stacking=True,
            cargo_group_name="Stackable"
        )

    used_all = used_unstackable + used_stackable

    # Renumber trucks after combining both groups.
    for index, result in enumerate(used_all, start=1):
        result["usage_number"] = index

    remaining_all = remaining_unstackable + remaining_stackable

    return used_all, remaining_all, 0.0


# ============================================================
# TEXT-BASED VISUALIZATION
# ============================================================

def print_3d_layers(container_width, container_depth, container_height, placements, main_box_name="Container"):
    """
    Prints the 3D container layer by layer.
    Each layer is a 2D map at a specific height z.
    Works best with whole number dimensions.
    """
    grid_width = int(container_width)
    grid_depth = int(container_depth)
    grid_height = int(container_height)

    grid = [
        [
            ["." for _ in range(grid_width)]
            for _ in range(grid_depth)
        ]
        for _ in range(grid_height)
    ]

    symbols = list("ABCDEFGHIJKLMNOPQRSTUVWXYZabcdefghijklmnopqrstuvwxyz0123456789")
    box_symbols = {}

    for index, placement in enumerate(placements):
        symbol = symbols[index] if index < len(symbols) else str(index + 1)
        box_symbols[placement.box.name] = symbol

        x_start = int(placement.x)
        y_start = int(placement.y)
        z_start = int(placement.z)

        # Iterate over the placed box's dimensions, clamped to container bounds.
        # This handles cases where gaps might push effective placement outside grid logic
        # (though ideally the packing algorithm prevents this for the box itself).
        for z in range(z_start, min(z_start + int(placement.height), grid_height)):
            for y in range(y_start, min(y_start + int(placement.depth), grid_depth)):
                for x in range(x_start, min(x_start + int(placement.width), grid_width)):
                    if 0 <= x < grid_width and 0 <= y < grid_depth and 0 <= z < grid_height:
                        grid[z][y][x] = symbol

    print(f"\n=== {main_box_name} 3D Container Layout by Height Layer ===")

    for z in range(grid_height):
        print(f"\nLayer z={z}")
        for y in range(grid_depth):
            print(" ".join(grid[z][y]))

    print("\nLegend:")
    for placement in placements:
        symbol = box_symbols[placement.box.name]
        print(
            f"{symbol} = {placement.box.name}: "
            f"{placement.width} x {placement.depth} x {placement.height}, "
            f"weight={placement.box.weight}, " # Added weight
            f"x={placement.x}, y={placement.y}, z={placement.z}, "
            f"rotated={placement.rotated}"
        )

    print(". = Empty space")


def create_box_mesh(x, y, z, w, d, h):
    """
    Creates the corner points and faces for a 3D rectangular box.
    This is used by Plotly Mesh3d.
    """
    vertices = [
        [x,     y,     z],        # 0
        [x + w, y,     z],        # 1
        [x + w, y + d, z],        # 2
        [x,     y + d, z],        # 3
        [x,     y,     z + h],    # 4
        [x + w, y,     z + h],    # 5
        [x + w, y + d, z + h],    # 6
        [x,     y + d, z + h],    # 7
    ]

    x_coords = [v[0] for v in vertices]
    y_coords = [v[1] for v in vertices]
    z_coords = [v[2] for v in vertices]

    # Define the 12 triangles that make up the 6 faces of the cube
    # Each face needs two triangles
    i = [0, 0, 0, 1, 1, 2, 4, 4, 4, 5, 5, 6]
    j = [1, 2, 4, 2, 5, 3, 5, 6, 7, 6, 1, 2]
    k = [2, 3, 5, 5, 6, 7, 6, 7, 0, 2, 0, 3]

    return x_coords, y_coords, z_coords, i, j, k


def visualize_3d_packing(
    container_width,
    container_depth,
    container_height,
    placements,
    title="3D Container Packing Visualization",
    show=True,
    camera=None
):
    """
    Builds a Plotly 3D figure.

    Parameters:
    - show=True: display the interactive figure in Colab/Jupyter.
    - show=False: build the same figure without displaying it, useful for PDF export.
    - camera: optional Plotly camera dictionary for a fixed PDF view.

    Returns:
        plotly.graph_objects.Figure
    """
    fig = go.Figure()

    colors = [
        "red", "green", "blue", "orange", "purple", "cyan",
        "magenta", "yellow", "brown", "pink", "lime", "teal",
        "gold", "navy", "coral", "olive"
    ]

    for index, placement in enumerate(placements):
        x = placement.x
        y = placement.y
        z = placement.z
        w = placement.width
        d = placement.depth
        h = placement.height

        x_coords, y_coords, z_coords, i, j, k = create_box_mesh(
            x, y, z, w, d, h
        )

        fig.add_trace(go.Mesh3d(
            x=x_coords,
            y=y_coords,
            z=z_coords,
            i=i,
            j=j,
            k=k,
            color=colors[index % len(colors)],
            opacity=0.75,
            name=placement.box.name,
            hovertext=(
                f"Item: {placement.box.name}<br>"
                f"Position: x={x:.2f}, y={y:.2f}, z={z:.2f}<br>"
                f"Size: {w:.2f} x {d:.2f} x {h:.2f}<br>"
                f"Original Size: "
                f"{placement.box.width:.2f} x "
                f"{placement.box.depth:.2f} x "
                f"{placement.box.height:.2f}<br>"
                f"Weight: {placement.box.weight:.2f}<br>"
                f"Rotated: {placement.rotated}"
            ),
            hoverinfo="text"
        ))

        add_box_outline(
            fig, x, y, z, w, d, h, placement.box.name
        )

    add_container_outline(
        fig,
        container_width,
        container_depth,
        container_height
    )

    scene_settings = dict(
        xaxis_title="Length",
        yaxis_title="Width",
        zaxis_title="Height",
        xaxis=dict(range=[0, container_width]),
        yaxis=dict(range=[0, container_depth]),
        zaxis=dict(range=[0, container_height]),
        aspectmode="data"
    )

    if camera is not None:
        scene_settings["camera"] = camera

    fig.update_layout(
        title=title,
        scene=scene_settings,
        width=950,
        height=750,
        showlegend=True,
        margin=dict(l=20, r=20, t=80, b=20)
    )

    if show:
        fig.show()

    return fig


def add_box_outline(fig, x, y, z, w, d, h, name):
    """
    Adds black edge lines around each placed box.
    """
    # Define corners of the box
    corners = {
        "000": (x,     y,     z),        # 0
        "100": (x + w, y,     z),        # 1
        "110": (x + w, y + d, z),        # 2
        "010": (x,     y + d, z),        # 3
        "001": (x,     y,     z + h),    # 4
        "101": (x + w, y,     z + h),    # 5
        "111": (x + w, y + d, z + h),    # 6
        "011": (x,     y + d, z + h),    # 7
    }

    # Define the 12 edges of a cube by connecting corners
    edges = [
        ("000", "100"), ("100", "110"), ("110", "010"), ("010", "000"), # Bottom face
        ("001", "101"), ("101", "111"), ("111", "011"), ("011", "001"), # Top face
        ("000", "001"), ("100", "101"), ("110", "111"), ("010", "011")  # Vertical edges
    ]

    edge_x = []
    edge_y = []
    edge_z = []

    for start, end in edges:
        x1, y1, z1 = corners[start]
        x2, y2, z2 = corners[end]

        edge_x.extend([x1, x2, None]) # Add None to break the line for scatter3d
        edge_y.extend([y1, y2, None])
        edge_z.extend([z1, z2, None])

    fig.add_trace(go.Scatter3d(
        x=edge_x,
        y=edge_y,
        z=edge_z,
        mode="lines",
        line=dict(width=3, color="black"),
        name=f"{name} outline",
        showlegend=False
    ))


def add_container_outline(fig, container_width, container_depth, container_height):
    """
    Adds an outline showing the total container/platform boundary.
    """
    x = 0
    y = 0
    z = 0
    w = container_width
    d = container_depth
    h = container_height

    corners = {
        "000": (x,     y,     z),
        "100": (x + w, y,     z),
        "110": (x + w, y + d, z),
        "010": (x,     y + d, z),
        "001": (x,     y,     z + h),
        "101": (x + w, y,     z + h),
        "111": (x + w, y + d, z + h),
        "011": (x,     y + d, z + h),
    }

    edges = [
        ("000", "100"), ("100", "110"), ("110", "010"), ("010", "000"),
        ("001", "101"), ("101", "111"), ("111", "011"), ("011", "001"),
        ("000", "001"), ("100", "101"), ("110", "111"), ("010", "011")
    ]

    edge_x = []
    edge_y = []
    edge_z = []

    for start, end in edges:
        x1, y1, z1 = corners[start]
        x2, y2, z2 = corners[end]

        edge_x.extend([x1, x2, None])
        edge_y.extend([y1, y2, None])
        edge_z.extend([z1, z2, None])

    fig.add_trace(go.Scatter3d(
        x=edge_x,
        y=edge_y,
        z=edge_z,
        mode="lines",
        line=dict(width=7, color="black"),
        name="Container Boundary"
    ))


def visualize_multiple_main_boxes_3d(used_main_boxes_3d):
    """
    Displays one interactive Plotly visualization for every truck used.

    Returns:
        list[plotly.graph_objects.Figure]
    """
    figures = []

    for result in used_main_boxes_3d:
        main_box = result["main_box"]

        cargo_group = result.get("cargo_group", "Cargo")
        stacking_label = (
            "Stacking Enabled"
            if result.get("allow_stacking", False)
            else "Floor Only"
        )

        title = (
            f"Main Box {result['usage_number']} "
            f"[{cargo_group} - {stacking_label}]: "
            f"{main_box.name} "
            f"({main_box.width} x "
            f"{main_box.depth} x "
            f"{main_box.height} cm)"
            f"<br>Boxes: {len(result['placements'])} | "
            f"Floor Utilization: "
            f"{result['floor_utilization']:.2f}% | "
            f"Weight Utilization: "
            f"{result['weight_utilization']:.2f}%"
        )

        figure = visualize_3d_packing(
            main_box.width,
            main_box.depth,
            main_box.height,
            result["placements"],
            title=title,
            show=True
        )
        figures.append(figure)

    return figures



# ============================================================
# INTERACTIVE HTML REPORT EXPORT
# ============================================================

def create_packing_report_html(
    used_main_boxes_3d,
    remaining_boxes,
    total_item_count,
    packed_count,
    item_gap,
    container_gap,
    center_items,
    output_filename="packing_report.html"
):
    """Creates a standalone interactive HTML packing report."""
    from html import escape

    output_path = Path(output_filename).resolve()
    truck_counts = Counter(
        (
            result.get("cargo_group", "Cargo"),
            result["main_box"].name,
            "Stacking Enabled" if result.get("allow_stacking", False) else "Floor Only"
        )
        for result in used_main_boxes_3d
    )

    parts = ["""<!DOCTYPE html>
<html lang="en"><head><meta charset="UTF-8">
<meta name="viewport" content="width=device-width, initial-scale=1.0">
<title>3D Packing Optimizer Report</title>
<style>
body{margin:0;padding:30px;background:#f4f6f8;color:#222;font-family:Arial,sans-serif}
.wrap{max-width:1400px;margin:auto}.card{background:white;border-radius:10px;padding:20px;margin-bottom:25px;box-shadow:0 2px 8px rgba(0,0,0,.08)}
h1{text-align:center}table{width:100%;border-collapse:collapse;margin-top:10px}th,td{border:1px solid #bbb;padding:8px;text-align:left;vertical-align:top}th{background:#d9eaf7}tr:nth-child(even){background:#f7f7f7}.help{background:#eef5ff;border-left:4px solid #3b78b4;padding:12px;margin-bottom:20px}.plot{min-height:720px}.warn{color:#b14b00;font-weight:bold}.good{color:#167c31;font-weight:bold}
</style></head><body><div class="wrap">"""]
    parts.append("<h1>Multi-Main-Box 3D Packing Report</h1>")
    parts.append(
        '<p style="text-align:center;color:#666">'
        f'Generated: {datetime.now().strftime("%d %B %Y, %H:%M")}</p>'
    )
    parts.append(
        '<div class="help">Drag to rotate. Use the mouse wheel to zoom. '
        'Hold Shift and drag to pan. Hover over cargo for details.</div>'
    )

    status = "All item boxes were packed." if not remaining_boxes else "Some item boxes could not be packed."
    status_class = "good" if not remaining_boxes else "warn"
    parts.append(f"""<div class="card"><h2>Summary Result</h2><table>
<tr><th>Summary</th><th>Result</th></tr>
<tr><td>Total item boxes entered</td><td>{total_item_count}</td></tr>
<tr><td>Total item boxes packed</td><td>{packed_count}</td></tr>
<tr><td>Total item boxes not packed</td><td>{len(remaining_boxes)}</td></tr>
<tr><td>Total main boxes used</td><td>{len(used_main_boxes_3d)}</td></tr>
<tr><td>Container clearance</td><td>{container_gap:.2f} cm</td></tr>
<tr><td>Item-to-item gap</td><td>{item_gap:.2f} cm</td></tr>
<tr><td>Center items</td><td>{center_items}</td></tr>
<tr><td>Packing status</td><td class="{status_class}">{status}</td></tr></table></div>""")

    parts.append('<div class="card"><h2>Truck Types Used</h2><table><tr><th>Amount</th><th>Truck type</th><th>Cargo group</th><th>Stacking mode</th></tr>')
    for (cargo_group, truck_name, stacking_label), count in truck_counts.items():
        parts.append(
            f'<tr><td>{count}</td><td>{escape(str(truck_name))}</td>'
            f'<td>{escape(str(cargo_group))}</td><td>{escape(stacking_label)}</td></tr>'
        )
    parts.append('</table></div>')

    if remaining_boxes:
        parts.append('<div class="card"><h2>Item Boxes That Could Not Fit</h2><table><tr><th>Item type</th><th>Length × Width × Height</th><th>Weight/item</th><th>Stacking</th><th>Amount</th></tr>')
        for (name, width, depth, height, weight, stackable), count in group_unpacked_boxes(remaining_boxes).items():
            parts.append(
                f'<tr><td>{escape(str(name))}</td>'
                f'<td>{width:.1f} × {depth:.1f} × {height:.1f} cm</td>'
                f'<td>{weight:.1f} kg</td>'
                f'<td>{"Stackable" if stackable else "Unstackable"}</td>'
                f'<td>{count}</td></tr>'
            )
        parts.append('</table></div>')

    include_js = True
    for result in used_main_boxes_3d:
        main_box = result["main_box"]
        placements = result["placements"]
        usage_number = result["usage_number"]
        cargo_group = result.get("cargo_group", "Cargo")
        stacking_label = "Stacking Enabled" if result.get("allow_stacking", False) else "Floor Only"

        figure = visualize_3d_packing(
            main_box.width,
            main_box.depth,
            main_box.height,
            placements,
            title=(
                f"Truck {usage_number}: {main_box.name}<br>"
                f"{cargo_group} — {stacking_label}<br>"
                f"Boxes: {len(placements)} | "
                f"Floor utilization: {result['floor_utilization']:.2f}% | "
                f"Weight utilization: {result['weight_utilization']:.2f}%"
            ),
            show=False
        )
        figure.update_layout(autosize=True, width=None, height=720)
        plot_html = pio.to_html(
            figure,
            full_html=False,
            include_plotlyjs=True if include_js else False,
            config={"responsive": True, "displaylogo": False, "scrollZoom": True}
        )
        include_js = False

        parts.append(f"""<div class="card"><h2>Used Main Box {usage_number}: {escape(main_box.name)}</h2><table>
<tr><th>Truck detail</th><th>Value</th></tr>
<tr><td>Cargo group</td><td>{escape(str(cargo_group))}</td></tr>
<tr><td>Stacking mode</td><td>{escape(stacking_label)}</td></tr>
<tr><td>Cargo dimensions</td><td>{main_box.width:.1f} × {main_box.depth:.1f} × {main_box.height:.1f} cm</td></tr>
<tr><td>Maximum weight</td><td>{main_box.max_weight:.2f} kg</td></tr>
<tr><td>Boxes packed</td><td>{len(placements)}</td></tr>
<tr><td>Floor utilization</td><td>{result['floor_utilization']:.2f}%</td></tr>
<tr><td>Weight utilization</td><td>{result['weight_utilization']:.2f}%</td></tr>
<tr><td>Volume utilization</td><td>{result['volume_utilization']:.2f}%</td></tr>
</table><div class="plot">{plot_html}</div>
<h3>Box Placements</h3><table><tr><th>Item</th><th>x</th><th>y</th><th>z</th><th>L</th><th>W</th><th>H</th><th>kg</th><th>Rotated</th><th>Stackable</th></tr>""")
        for placement in placements:
            parts.append(
                f'<tr><td>{escape(placement.box.name)}</td>'
                f'<td>{placement.x:.1f}</td><td>{placement.y:.1f}</td><td>{placement.z:.1f}</td>'
                f'<td>{placement.width:.1f}</td><td>{placement.depth:.1f}</td><td>{placement.height:.1f}</td>'
                f'<td>{placement.box.weight:.1f}</td>'
                f'<td>{"Yes" if placement.rotated else "No"}</td>'
                f'<td>{"Yes" if placement.box.stackable else "No"}</td></tr>'
            )
        parts.append('</table></div>')

    parts.append('</div></body></html>')
    output_path.write_text(''.join(parts), encoding='utf-8')
    if not output_path.exists() or output_path.stat().st_size == 0:
        raise RuntimeError("The HTML report was not created correctly.")
    return str(output_path)


# ============================================================
# ROBUST STATIC IMAGE EXPORT HELPERS
# ============================================================

def configure_kaleido_browser():
    """
    Tries to locate Chrome/Chromium and exposes it to Kaleido through
    the BROWSER_PATH environment variable.

    The commands below should still be run once in Google Colab:

        !pip install -q -U plotly kaleido reportlab
        !plotly_get_chrome -y

    Returns:
        str | None: detected browser path, or None if no browser was found.
    """
    existing = os.environ.get("BROWSER_PATH")
    if existing and Path(existing).exists():
        return existing

    executable_names = [
        "google-chrome",
        "google-chrome-stable",
        "chromium",
        "chromium-browser",
        "chrome",
    ]

    for executable_name in executable_names:
        detected = shutil.which(executable_name)
        if detected:
            os.environ["BROWSER_PATH"] = detected
            return detected

    common_paths = [
        "/usr/bin/google-chrome",
        "/usr/bin/google-chrome-stable",
        "/usr/bin/chromium",
        "/usr/bin/chromium-browser",
        "/opt/google/chrome/chrome",
    ]

    for candidate in common_paths:
        if Path(candidate).exists():
            os.environ["BROWSER_PATH"] = candidate
            return candidate

    return None


def export_plotly_figure_to_png(
    figure,
    image_path,
    width=1300,
    height=850,
    scale=1
):
    """
    Exports a Plotly figure to PNG using Kaleido.

    This function:
    - explicitly detects Chrome/Chromium,
    - uses plotly.io.to_image instead of hiding the real exception,
    - verifies that the resulting PNG is non-empty.

    Raises:
        RuntimeError: with the real Plotly/Kaleido error.
    """
    browser_path = configure_kaleido_browser()

    try:
        image_bytes = pio.to_image(
            figure,
            format="png",
            width=width,
            height=height,
            scale=scale
        )

        if not image_bytes:
            raise RuntimeError("Kaleido returned empty image data.")

        image_path = Path(image_path)
        image_path.write_bytes(image_bytes)

        if not image_path.exists() or image_path.stat().st_size == 0:
            raise RuntimeError("The PNG file was not created correctly.")

        return str(image_path)

    except Exception as exc:
        browser_message = (
            browser_path
            if browser_path
            else "Chrome/Chromium was not detected"
        )

        raise RuntimeError(
            "Plotly/Kaleido static image export failed.\n"
            f"Detected browser: {browser_message}\n"
            f"Actual error type: {type(exc).__name__}\n"
            f"Actual error message: {exc}"
        ) from exc


def save_matplotlib_truck_image(result, image_path):
    """
    Fallback PDF image renderer.

    If Plotly/Kaleido still fails in Colab, this creates a static 3D truck
    image with Matplotlib. The interactive Plotly figures remain unchanged
    in the notebook; only the PDF snapshot uses this fallback.
    """
    try:
        import matplotlib.pyplot as plt
        from matplotlib.patches import Patch
    except ImportError as exc:
        raise RuntimeError(
            "Matplotlib fallback is unavailable. "
            "Install it with: !pip install -q matplotlib"
        ) from exc

    main_box = result["main_box"]
    placements = result["placements"]

    fig = plt.figure(figsize=(12, 8))
    ax = fig.add_subplot(111, projection="3d")

    color_cycle = list(plt.cm.tab20.colors)
    legend_handles = []

    for index, placement in enumerate(placements):
        color = color_cycle[index % len(color_cycle)]

        ax.bar3d(
            placement.x,
            placement.y,
            placement.z,
            placement.width,
            placement.depth,
            placement.height,
            color=color,
            edgecolor="black",
            linewidth=0.6,
            alpha=0.75,
            shade=True
        )

        legend_handles.append(
            Patch(
                facecolor=color,
                edgecolor="black",
                label=placement.box.name
            )
        )

    # Container boundary.
    x0, y0, z0 = 0.0, 0.0, 0.0
    x1 = main_box.width
    y1 = main_box.depth
    z1 = main_box.height

    corners = [
        (x0, y0, z0), (x1, y0, z0), (x1, y1, z0), (x0, y1, z0),
        (x0, y0, z1), (x1, y0, z1), (x1, y1, z1), (x0, y1, z1),
    ]

    edges = [
        (0, 1), (1, 2), (2, 3), (3, 0),
        (4, 5), (5, 6), (6, 7), (7, 4),
        (0, 4), (1, 5), (2, 6), (3, 7),
    ]

    for start_index, end_index in edges:
        start = corners[start_index]
        end = corners[end_index]

        ax.plot(
            [start[0], end[0]],
            [start[1], end[1]],
            [start[2], end[2]],
            color="black",
            linewidth=1.5
        )

    ax.set_xlim(0, main_box.width)
    ax.set_ylim(0, main_box.depth)
    ax.set_zlim(0, main_box.height)

    ax.set_xlabel("Length (cm)")
    ax.set_ylabel("Width (cm)")
    ax.set_zlabel("Height (cm)")

    ax.set_title(
        f"Truck {result['usage_number']}: {main_box.name}\n"
        f"{result.get('cargo_group', 'Cargo')} - "
        f"{'Stacking Enabled' if result.get('allow_stacking', False) else 'Floor Only'} | "
        f"Boxes: {len(placements)} | "
        f"Floor utilization: {result['floor_utilization']:.2f}% | "
        f"Weight utilization: {result['weight_utilization']:.2f}%"
    )

    ax.view_init(elev=24, azim=-55)

    # Keep the legend manageable for large runs.
    if legend_handles:
        maximum_legend_items = 16
        handles_to_show = legend_handles[:maximum_legend_items]

        if len(legend_handles) > maximum_legend_items:
            handles_to_show.append(
                Patch(
                    facecolor="white",
                    edgecolor="white",
                    label=(
                        f"... and "
                        f"{len(legend_handles) - maximum_legend_items} more"
                    )
                )
            )

        ax.legend(
            handles=handles_to_show,
            loc="upper left",
            bbox_to_anchor=(1.02, 1.0),
            fontsize=7
        )

    fig.tight_layout()
    fig.savefig(
        image_path,
        format="png",
        dpi=160,
        bbox_inches="tight"
    )
    plt.close(fig)

    image_path = Path(image_path)

    if not image_path.exists() or image_path.stat().st_size == 0:
        raise RuntimeError(
            "Matplotlib fallback did not create a valid PNG file."
        )

    return str(image_path)


# ============================================================
# A4 PDF REPORT EXPORT
# ============================================================

def get_base_item_name(item_name):
    """
    Removes the generated numeric suffix from names such as:
    'Laptop Box-12' -> 'Laptop Box'
    """
    return re.sub(r"-\d+$", "", item_name)


def group_unpacked_boxes(remaining_boxes):
    """
    Groups failed items by base name, dimensions, and weight.
    """
    grouped = Counter()

    for box in remaining_boxes:
        key = (
            get_base_item_name(box.name),
            box.width,
            box.depth,
            box.height,
            box.weight,
            box.stackable
        )
        grouped[key] += 1

    return grouped


def make_report_figure(result):
    """
    Builds a fixed isometric view for inclusion in the PDF.
    The normal interactive figure remains available in Colab.
    """
    main_box = result["main_box"]

    camera = dict(
        eye=dict(x=1.55, y=1.55, z=1.15),
        center=dict(x=0, y=0, z=-0.1),
        up=dict(x=0, y=0, z=1)
    )

    title = (
        f"Truck {result['usage_number']}: {main_box.name}<br>"
        f"{result.get('cargo_group', 'Cargo')} - "
        f"{'Stacking Enabled' if result.get('allow_stacking', False) else 'Floor Only'} | "
        f"Boxes: {len(result['placements'])} | "
        f"Floor utilization: {result['floor_utilization']:.2f}% | "
        f"Weight utilization: {result['weight_utilization']:.2f}%"
    )

    return visualize_3d_packing(
        main_box.width,
        main_box.depth,
        main_box.height,
        result["placements"],
        title=title,
        show=False,
        camera=camera
    )


def create_packing_report_pdf(
    used_main_boxes_3d,
    remaining_boxes,
    total_item_count,
    packed_count,
    item_gap,
    container_gap,
    center_items,
    output_filename="packing_report.pdf"
):
    """
    Creates a multi-page A4 PDF containing:
    - the overall packing summary,
    - the grouped truck breakdown,
    - the grouped unpacked-item breakdown,
    - one detailed section and static 3D image per used truck,
    - placement tables for each truck.

    Plotly remains interactive in the notebook, while the PDF contains
    static snapshots because PDF does not preserve Plotly interaction.
    """
    try:
        from reportlab.lib import colors
        from reportlab.lib.enums import TA_CENTER
        from reportlab.lib.pagesizes import A4
        from reportlab.lib.styles import getSampleStyleSheet, ParagraphStyle
        from reportlab.lib.units import cm
        from reportlab.platypus import (
            SimpleDocTemplate,
            Paragraph,
            Spacer,
            Image,
            Table,
            TableStyle,
            PageBreak,
            LongTable,
            KeepTogether
        )
    except ImportError as exc:
        raise RuntimeError(
            "ReportLab is not installed. In Google Colab, run:\\n"
            "!pip install -q reportlab kaleido"
        ) from exc

    output_path = Path(output_filename).resolve()
    styles = getSampleStyleSheet()

    title_style = ParagraphStyle(
        "ReportTitle",
        parent=styles["Title"],
        alignment=TA_CENTER,
        fontSize=18,
        leading=22,
        spaceAfter=14
    )

    section_style = ParagraphStyle(
        "SectionHeading",
        parent=styles["Heading2"],
        fontSize=13,
        leading=16,
        spaceBefore=8,
        spaceAfter=8
    )

    small_style = ParagraphStyle(
        "SmallText",
        parent=styles["BodyText"],
        fontSize=8,
        leading=10
    )

    normal_style = styles["BodyText"]

    doc = SimpleDocTemplate(
        str(output_path),
        pagesize=A4,
        rightMargin=1.4 * cm,
        leftMargin=1.4 * cm,
        topMargin=1.4 * cm,
        bottomMargin=1.4 * cm,
        title="3D Packing Optimizer Report",
        author="3D Packing Optimizer"
    )

    story = []

    story.append(Paragraph(
        "Multi-Main-Box 3D Packing Report",
        title_style
    ))
    story.append(Paragraph(
        f"Generated: {datetime.now().strftime('%d %B %Y, %H:%M')}",
        normal_style
    ))
    story.append(Spacer(1, 8))

    truck_counts = Counter(
        (
            result.get("cargo_group", "Cargo"),
            result["main_box"].name,
            "Stacking Enabled" if result.get("allow_stacking", False) else "Floor Only"
        )
        for result in used_main_boxes_3d
    )

    summary_data = [
        ["Summary", "Result"],
        ["Total item boxes entered", str(total_item_count)],
        ["Total item boxes packed", str(packed_count)],
        ["Total item boxes not packed", str(len(remaining_boxes))],
        ["Total main boxes used", str(len(used_main_boxes_3d))],
        ["Container clearance", f"{container_gap:.2f} cm"],
        ["Item-to-item gap", f"{item_gap:.2f} cm"],
        ["Center items", str(center_items)],
        [
            "Packing status",
            (
                "All item boxes were packed."
                if not remaining_boxes
                else "Some item boxes could not be packed."
            )
        ]
    ]

    summary_table = Table(
        summary_data,
        colWidths=[7.0 * cm, 10.0 * cm],
        repeatRows=1
    )
    summary_table.setStyle(TableStyle([
        ("BACKGROUND", (0, 0), (-1, 0), colors.HexColor("#D9EAF7")),
        ("TEXTCOLOR", (0, 0), (-1, 0), colors.black),
        ("FONTNAME", (0, 0), (-1, 0), "Helvetica-Bold"),
        ("FONTNAME", (0, 1), (0, -1), "Helvetica-Bold"),
        ("GRID", (0, 0), (-1, -1), 0.5, colors.grey),
        ("VALIGN", (0, 0), (-1, -1), "TOP"),
        ("ROWBACKGROUNDS", (0, 1), (-1, -1), [
            colors.white,
            colors.HexColor("#F5F7F9")
        ]),
        ("FONTSIZE", (0, 0), (-1, -1), 9),
        ("LEFTPADDING", (0, 0), (-1, -1), 6),
        ("RIGHTPADDING", (0, 0), (-1, -1), 6),
        ("TOPPADDING", (0, 0), (-1, -1), 5),
        ("BOTTOMPADDING", (0, 0), (-1, -1), 5)
    ]))
    story.append(summary_table)
    story.append(Spacer(1, 14))

    story.append(Paragraph("Truck Types Used", section_style))

    if truck_counts:
        truck_data = [["Amount", "Truck type", "Cargo group", "Stacking mode"]]
        for (cargo_group, truck_name, stacking_label), count in truck_counts.items():
            truck_data.append([str(count), truck_name, cargo_group, stacking_label])

        truck_table = Table(
            truck_data,
            colWidths=[2.0 * cm, 7.0 * cm, 4.0 * cm, 4.0 * cm],
            repeatRows=1
        )
        truck_table.setStyle(TableStyle([
            ("BACKGROUND", (0, 0), (-1, 0), colors.HexColor("#E4F1E7")),
            ("FONTNAME", (0, 0), (-1, 0), "Helvetica-Bold"),
            ("GRID", (0, 0), (-1, -1), 0.5, colors.grey),
            ("VALIGN", (0, 0), (-1, -1), "TOP"),
            ("FONTSIZE", (0, 0), (-1, -1), 9),
            ("ALIGN", (0, 1), (0, -1), "CENTER")
        ]))
        story.append(truck_table)
    else:
        story.append(Paragraph(
            "No truck was used because no item could be packed.",
            normal_style
        ))

    if remaining_boxes:
        story.append(Spacer(1, 14))
        story.append(Paragraph(
            "Item Boxes That Could Not Fit",
            section_style
        ))

        unpacked_data = [[
            "Item type",
            "Length × Width × Height",
            "Weight/item",
            "Stacking",
            "Amount"
        ]]

        grouped_unpacked = group_unpacked_boxes(remaining_boxes)

        for (
            name,
            width,
            depth,
            height,
            weight,
            stackable
        ), count in grouped_unpacked.items():
            unpacked_data.append([
                name,
                f"{width:.1f} × {depth:.1f} × {height:.1f} cm",
                f"{weight:.1f} kg",
                "Stackable" if stackable else "Unstackable",
                str(count)
            ])

        unpacked_table = LongTable(
            unpacked_data,
            colWidths=[
                4.5 * cm,
                5.8 * cm,
                2.5 * cm,
                2.6 * cm,
                1.6 * cm
            ],
            repeatRows=1
        )
        unpacked_table.setStyle(TableStyle([
            ("BACKGROUND", (0, 0), (-1, 0), colors.HexColor("#F7DFDF")),
            ("FONTNAME", (0, 0), (-1, 0), "Helvetica-Bold"),
            ("GRID", (0, 0), (-1, -1), 0.5, colors.grey),
            ("VALIGN", (0, 0), (-1, -1), "TOP"),
            ("FONTSIZE", (0, 0), (-1, -1), 8),
            ("ALIGN", (-1, 1), (-1, -1), "CENTER")
        ]))
        story.append(unpacked_table)

    with tempfile.TemporaryDirectory() as temp_dir:
        temp_path = Path(temp_dir)

        for result in used_main_boxes_3d:
            main_box = result["main_box"]
            usage_number = result["usage_number"]

            story.append(PageBreak())
            story.append(Paragraph(
                f"Used Main Box {usage_number}: {main_box.name}",
                section_style
            ))

            detail_data = [
                ["Truck detail", "Value"],
                ["Cargo group", result.get("cargo_group", "Cargo")],
                [
                    "Stacking mode",
                    (
                        "Stacking Enabled"
                        if result.get("allow_stacking", False)
                        else "Floor Only"
                    )
                ],
                [
                    "Cargo dimensions",
                    (
                        f"{main_box.width:.1f} × "
                        f"{main_box.depth:.1f} × "
                        f"{main_box.height:.1f} cm"
                    )
                ],
                ["Maximum weight", f"{main_box.max_weight:.2f} kg"],
                ["Boxes packed", str(len(result["placements"]))],
                ["Used floor area", f"{result['fitted_floor_area']:.2f} cm²"],
                [
                    "Available floor area",
                    f"{main_box.width * main_box.depth:.2f} cm²"
                ],
                [
                    "Floor utilization",
                    f"{result['floor_utilization']:.2f}%"
                ],
                ["Packed weight", f"{result['fitted_weight']:.2f} kg"],
                [
                    "Weight utilization",
                    f"{result['weight_utilization']:.2f}%"
                ],
                ["Used volume", f"{result['fitted_volume']:.2f} cm³"],
                [
                    "Volume utilization",
                    f"{result['volume_utilization']:.2f}%"
                ]
            ]

            detail_table = Table(
                detail_data,
                colWidths=[6.0 * cm, 11.0 * cm],
                repeatRows=1
            )
            detail_table.setStyle(TableStyle([
                ("BACKGROUND", (0, 0), (-1, 0), colors.HexColor("#D9EAF7")),
                ("FONTNAME", (0, 0), (-1, 0), "Helvetica-Bold"),
                ("FONTNAME", (0, 1), (0, -1), "Helvetica-Bold"),
                ("GRID", (0, 0), (-1, -1), 0.5, colors.grey),
                ("VALIGN", (0, 0), (-1, -1), "TOP"),
                ("FONTSIZE", (0, 0), (-1, -1), 8.5)
            ]))
            story.append(detail_table)
            story.append(Spacer(1, 10))

            figure = make_report_figure(result)
            image_path = temp_path / f"truck_{usage_number}.png"

            try:
                export_plotly_figure_to_png(
                    figure=figure,
                    image_path=image_path,
                    width=1300,
                    height=850,
                    scale=1
                )

            except Exception as kaleido_error:
                print(
                    "\nPlotly/Kaleido could not create the PDF image. "
                    "Using the Matplotlib fallback instead."
                )
                print(kaleido_error)

                try:
                    save_matplotlib_truck_image(
                        result=result,
                        image_path=image_path
                    )

                except Exception as fallback_error:
                    raise RuntimeError(
                        "Both static-image export methods failed.\n\n"
                        "PLOTLY/KALEIDO ERROR:\n"
                        f"{kaleido_error}\n\n"
                        "MATPLOTLIB FALLBACK ERROR:\n"
                        f"{fallback_error}\n\n"
                        "Full fallback traceback:\n"
                        f"{traceback.format_exc()}"
                    ) from fallback_error

            report_image = Image(
                str(image_path),
                width=17.0 * cm,
                height=11.1 * cm
            )
            story.append(report_image)
            story.append(Spacer(1, 10))

            story.append(Paragraph("Box Placements", section_style))

            placement_data = [[
                "Item",
                "x",
                "y",
                "z",
                "L",
                "W",
                "H",
                "kg",
                "Stack",
                "Rot."
            ]]

            for placement in result["placements"]:
                placement_data.append([
                    Paragraph(placement.box.name, small_style),
                    f"{placement.x:.1f}",
                    f"{placement.y:.1f}",
                    f"{placement.z:.1f}",
                    f"{placement.width:.1f}",
                    f"{placement.depth:.1f}",
                    f"{placement.height:.1f}",
                    f"{placement.box.weight:.1f}",
                    "Yes" if placement.box.stackable else "No",
                    "Yes" if placement.rotated else "No"
                ])

            placement_table = LongTable(
                placement_data,
                colWidths=[
                    4.1 * cm,
                    1.25 * cm,
                    1.25 * cm,
                    1.25 * cm,
                    1.25 * cm,
                    1.25 * cm,
                    1.25 * cm,
                    1.35 * cm,
                    1.35 * cm,
                    1.15 * cm
                ],
                repeatRows=1
            )
            placement_table.setStyle(TableStyle([
                ("BACKGROUND", (0, 0), (-1, 0), colors.HexColor("#EAEAEA")),
                ("FONTNAME", (0, 0), (-1, 0), "Helvetica-Bold"),
                ("GRID", (0, 0), (-1, -1), 0.35, colors.grey),
                ("VALIGN", (0, 0), (-1, -1), "MIDDLE"),
                ("ALIGN", (1, 1), (-1, -1), "CENTER"),
                ("FONTSIZE", (0, 0), (-1, -1), 7),
                ("TOPPADDING", (0, 0), (-1, -1), 3),
                ("BOTTOMPADDING", (0, 0), (-1, -1), 3)
            ]))
            story.append(placement_table)

        doc.build(story)

    return str(output_path)


def download_file_in_colab(file_path):
    """
    Starts a browser download when running in Google Colab.
    Outside Colab, it prints the saved file location.
    """
    try:
        from google.colab import files
        files.download(file_path)
    except ImportError:
        print(f"PDF saved to: {file_path}")


# ============================================================
# CSV / EXCEL ITEM INPUT
# ============================================================

def load_item_boxes_from_file(file_path):
    """
    Reads item-box data from a CSV or Excel file.

    Required columns (case-insensitive):
    - Name
    - Length
    - Width
    - Height
    - Quantity
    - Weight
    - Stackable

    Dimensions must be in centimeters and Weight must be in kilograms
    per item. Each row is expanded into individual Box3D objects.
    """
    file_path = Path(file_path)

    if not file_path.exists():
        raise FileNotFoundError(f"Input file was not found: {file_path}")

    extension = file_path.suffix.lower()

    if extension == ".csv":
        dataframe = pd.read_csv(file_path)
    elif extension in [".xlsx", ".xls"]:
        dataframe = pd.read_excel(file_path)
    else:
        raise ValueError("Unsupported file type. Use CSV, XLSX, or XLS.")

    # Normalize headings so NAME, Name, name, and headings with spaces work.
    normalized_columns = {
        str(column).strip().lower(): column
        for column in dataframe.columns
    }

    required_columns = [
        "name", "length", "width", "height",
        "quantity", "weight", "stackable"
    ]

    missing_columns = [
        column for column in required_columns
        if column not in normalized_columns
    ]

    if missing_columns:
        raise ValueError(
            "The file is missing these required columns: "
            + ", ".join(column.title() for column in missing_columns)
        )

    item_boxes_3d = []

    for dataframe_index, row in dataframe.iterrows():
        source_row_number = dataframe_index + 2

        # Completely empty rows are ignored.
        if row.isna().all():
            continue

        try:
            name_value = row[normalized_columns["name"]]
            name = "" if pd.isna(name_value) else str(name_value).strip()
            if not name:
                name = f"ItemBoxRow{source_row_number}"

            length = float(row[normalized_columns["length"]])
            width = float(row[normalized_columns["width"]])
            height = float(row[normalized_columns["height"]])
            quantity_value = float(row[normalized_columns["quantity"]])
            weight = float(row[normalized_columns["weight"]])
            stackable = parse_stackable_value(
                row[normalized_columns["stackable"]]
            )

            if not quantity_value.is_integer():
                raise ValueError("Quantity must be a whole number.")

            quantity = int(quantity_value)

            if length <= 0:
                raise ValueError("Length must be greater than zero.")
            if width <= 0:
                raise ValueError("Width must be greater than zero.")
            if height <= 0:
                raise ValueError("Height must be greater than zero.")
            if quantity <= 0:
                raise ValueError("Quantity must be greater than zero.")
            if weight <= 0:
                raise ValueError("Weight must be greater than zero.")

        except (TypeError, ValueError) as error:
            raise ValueError(
                f"Invalid data in CSV/Excel row {source_row_number}: {error}"
            ) from error

        for item_number in range(1, quantity + 1):
            item_boxes_3d.append(
                Box3D(
                    name=f"{name}-{item_number}",
                    width=length,
                    depth=width,
                    height=height,
                    weight=weight,
                    stackable=stackable
                )
            )

    if not item_boxes_3d:
        raise ValueError("The uploaded file does not contain any valid item boxes.")

    return item_boxes_3d


# ============================================================
# INPUT VALIDATION HELPERS (Copied from 2D version)
# ============================================================

def get_positive_float(prompt):
    """
    Repeatedly asks for a positive number.
    """

    while True:
        try:
            value = float(input(prompt))

            if value <= 0:
                print("Please enter a number greater than zero.")
                continue

            return value

        except ValueError:
            print("Invalid input. Please enter a number.")

def get_positive_or_zero_float(prompt):
    """
    Repeatedly asks for a number greater than or equal to zero.
    A value of 0 means no gap.
    """

    while True:
        try:
            value = float(input(prompt))

            if value < 0:
                print("Please enter zero or a positive number.")
                continue

            return value

        except ValueError:
            print("Invalid input. Please enter a number.")


def get_positive_integer(prompt):
    """
    Repeatedly asks for a positive whole number.
    """

    while True:
        try:
            value = int(input(prompt))

            if value <= 0:
                print("Please enter a whole number greater than zero.")
                continue

            return value

        except ValueError:
            print("Invalid input. Please enter a whole number.")


def get_yes_no(prompt):
    """
    Repeatedly asks until the user enters yes or no.
    """

    while True:
        answer = input(prompt).strip().lower()

        if answer in ["yes", "y"]:
            return True

        if answer in ["no", "n"]:
            return False

        print("Please enter yes or no.")


# ============================================================
# MAIN PROGRAM (3D)
# ============================================================

def main_3d():
    print("=== Multi-Main-Box 3D Packing Optimizer ===")

    # --------------------------------------------------------
    # INPUT MAIN-BOX TYPES
    # --------------------------------------------------------

    main_boxes_3d = []

    print("\n=== Using Default Main Box Types from CSV (Unscaled values for display) ===")
    # Predefined Indonesian truck specifications in centimeters
    predefined_truck_specs = [
        {"name": "Pick Up (Bak/Box)", "width": 230.0, "depth": 140.0, "height": 130.0, "max_weight": 1000},
        {"name": "Colt Diesel Engkel (CDE)", "width": 310.0, "depth": 170.0, "height": 170.0, "max_weight": 2000},
        {"name": "Colt Diesel Double (CDD)", "width": 450.0, "depth": 200.0, "height": 200.0, "max_weight": 5000},
        {"name": "CDD Long", "width": 550.0, "depth": 220.0, "height": 220.0, "max_weight": 5000},
        {"name": "Truk Fuso (Engkel Medium)", "width": 600.0, "depth": 250.0, "height": 280.0, "max_weight": 8000},
        {"name": "Truk Tronton", "width": 900.0, "depth": 250.0, "height": 280.0, "max_weight": 15000},
        {"name": "Tronton Wingbox", "width": 900.0, "depth": 250.0, "height": 250.0, "max_weight": 15000},
        {"name": "Trailer Truck 20 Feet", "width": 600.0, "depth": 250.0, "height": 280.0, "max_weight": 25000},
        {"name": "Trailer Truck 40 Feet", "width": 1200.0, "depth": 250.0, "height": 280.0, "max_weight": 25000},
        {"name": "Trailer Lowbet", "width": 700.0, "depth": 300.0, "height": 350.0, "max_weight": 35000},
    ]

    print("All available truck types will be used:")
    for i, truck in enumerate(predefined_truck_specs):
        print(f"  - {truck['name']}: {truck['width']}x{truck['depth']}x{truck['height']} cm, Max Weight: {truck['max_weight']} kg") # Removed score from print

    # Automatically add all predefined truck types
    for truck_data in predefined_truck_specs:
        main_boxes_3d.append(
            MainBox3D(
                name=truck_data["name"],
                width=truck_data["width"],
                depth=truck_data["depth"],
                height=truck_data["height"],
                # score=truck_data["score"],
                max_weight=truck_data["max_weight"]
            )
        )
    print(f"Added {len(main_boxes_3d)} truck types for consideration.")

    # --------------------------------------------------------
    # INPUT ITEM-BOX TYPES
    # --------------------------------------------------------

    item_boxes_3d = []

    print("\n=== Item Box Input Method ===")
    print("1. Enter item boxes manually")
    print("2. Upload a CSV or Excel file")

    while True:
        input_method = input("Choose input method (1 or 2): ").strip()

        if input_method in ["1", "2"]:
            break

        print("Please enter 1 or 2.")

    # --------------------------------------------------------
    # OPTION 1: MANUAL INPUT
    # --------------------------------------------------------

    if input_method == "1":
        item_box_type_number = 1

        print("\n=== Enter Item 3D Boxes Manually ===")

        while True:
            print(f"\nItem Box Type {item_box_type_number}")

            name = input(
                "Enter item box name (e.g., 'Laptop Box'): "
            ).strip()

            if not name:
                name = f"ItemBoxType{item_box_type_number}"

            length = get_positive_float(
                "Enter item box length (cm): "
            )
            width = get_positive_float(
                "Enter item box width (cm): "
            )
            height = get_positive_float(
                "Enter item box height (cm): "
            )
            quantity = get_positive_integer(
                "Enter item box quantity: "
            )
            weight = get_positive_float(
                "Enter weight of each item box (kg): "
            )

            stackable = get_yes_no(
                "Is this item type stackable? yes/no: "
            )

            for item_number in range(1, quantity + 1):
                item_boxes_3d.append(
                    Box3D(
                        name=f"{name}-{item_number}",
                        width=length,
                        depth=width,
                        height=height,
                        weight=weight,
                        stackable=stackable
                    )
                )

            add_more = get_yes_no(
                "\nAdd another item box type? yes/no: "
            )

            if not add_more:
                break

            item_box_type_number += 1

    # --------------------------------------------------------
    # OPTION 2: CSV OR EXCEL UPLOAD
    # --------------------------------------------------------

    else:
        print("\n=== Upload CSV or Excel Item File ===")
        print(
            "Required columns: Name, Length, Width, Height, "
            "Quantity, Weight, Stackable"
        )
        print(
            "Dimensions must be in cm and Weight must be kg per item."
        )

        try:
            from google.colab import files

            uploaded_files = files.upload()

            if not uploaded_files:
                raise ValueError("No file was uploaded.")

            uploaded_filename = next(iter(uploaded_files))
            item_boxes_3d = load_item_boxes_from_file(uploaded_filename)

        except ImportError:
            file_path = input(
                "Enter the CSV or Excel file path: "
            ).strip()
            item_boxes_3d = load_item_boxes_from_file(file_path)

        print(
            f"Successfully loaded {len(item_boxes_3d)} "
            "individual item boxes."
        )

    allow_rotation = get_yes_no(
        "\nAllow item box rotation? yes/no: "
    )

    container_gap = get_positive_or_zero_float(
        "\nEnter required gap between item boxes and main container (cm): "
    )

    item_gap = get_positive_or_zero_float(
        "\nEnter required gap between item boxes (cm): "
    )

    center_items = get_yes_no(
        "\nCenter all packed items in the main container? yes/no: "
    )

    # --------------------------------------------------------
    # RUN MULTIPLE MAIN-BOX PACKING (3D)
    # --------------------------------------------------------

    used_main_boxes_3d, remaining_boxes, total_score = (
        pack_separated_stackable_and_unstackable_items(
            main_boxes_3d=main_boxes_3d,
            item_boxes_3d=item_boxes_3d,
            allow_rotation=allow_rotation,
            item_gap=item_gap,
            container_gap=container_gap,
            center_items=center_items
        )
    )

    # --------------------------------------------------------
    # PRINT OVERALL RESULT
    # --------------------------------------------------------

    packed_count = sum(
        len(result["placements"])
        for result in used_main_boxes_3d
    )

    print("\n" + "=" * 55)
    print("MULTI-MAIN-BOX 3D PACKING RESULT")
    print("=" * 55)

    stackable_count = sum(1 for box in item_boxes_3d if box.stackable)
    unstackable_count = len(item_boxes_3d) - stackable_count

    print(f"Total item boxes entered: {len(item_boxes_3d)}")
    print(f"Stackable item boxes entered: {stackable_count}")
    print(f"Unstackable item boxes entered: {unstackable_count}")
    print(f"Total item boxes packed: {packed_count}")
    print(f"Total item boxes not packed: {len(remaining_boxes)}")
    print(f"Total main boxes used: {len(used_main_boxes_3d)}")

    # Add breakdown of used main boxes
    if used_main_boxes_3d:
        truck_counts = Counter(
            (
                result.get("cargo_group", "Cargo"),
                result["main_box"].name,
                "Stacking Enabled" if result.get("allow_stacking", False) else "Floor Only"
            )
            for result in used_main_boxes_3d
        )
        print("\nMain boxes used:")
        for (cargo_group, truck_name, stacking_label), count in truck_counts.items():
            print(f"- {count} × {truck_name} [{cargo_group} - {stacking_label}]")

    print(f"\nRequired gap between item boxes and main container: {container_gap} cm")
    print(f"Required gap between item boxes: {item_gap} cm")
    print(f"Center items: {center_items}")
    # print(f"Total score: {total_score:.2f}") # Removed total score print

    if not remaining_boxes:
        print("Packing status: All item boxes were packed.")
    else:
        print("Packing status: Some item boxes could not be packed.")

    # --------------------------------------------------------
    # PRINT EACH USED MAIN BOX
    # --------------------------------------------------------

    for result in used_main_boxes_3d:
        main_box = result["main_box"]
        placements = result["placements"]

        print("\n" + "-" * 55)
        print(f"USED MAIN BOX {result['usage_number']}")
        print("-" * 55)

        print(f"Cargo group: {result.get('cargo_group', 'Cargo')}")
        print(
            "Stacking mode: "
            + ("Stacking Enabled" if result.get("allow_stacking", False) else "Floor Only")
        )
        print(f"Type: {main_box.name}")
        print(f"Size: {main_box.width} x {main_box.depth} x {main_box.height} cm") # Added depth
        # print(f"Score: {main_box.score}") # Removed score print
        print(f"Maximum weight: {main_box.max_weight} kg")
        print(f"Boxes packed here: {len(placements)}")
        print(f"Used volume: {result['fitted_volume']:.2f} cm³") # Changed to volume
        print(f"Main-box volume: {main_box.volume():.2f} cm³") # Changed to volume
        print(
            f"Used floor area: "
            f"{result['fitted_floor_area']:.2f} cm²"
        )

        print(
            f"Main-box floor area: "
            f"{main_box.width * main_box.depth:.2f} cm²"
        )

        print(
            f"Floor utilization: "
            f"{result['floor_utilization']:.2f}%"
        )

        print(
            f"Volume utilization: "
            f"{result['volume_utilization']:.2f}%"
        )
        print(f"Packed weight: {result['fitted_weight']:.2f} kg")
        print(f"Weight utilization: {result['weight_utilization']:.2f} %\n")

        print("Box placements:")

        for placement in placements:
            print(
                f"{placement.box.name}: "
                f"x={placement.x:.2f} cm, "
                f"y={placement.y:.2f} cm, "
                f"z={placement.z:.2f} cm, " # Added z
                f"w={placement.width:.2f} cm, "
                f"d={placement.depth:.2f} cm, " # Added depth
                f"h={placement.height:.2f} cm, "
                f"weight={placement.box.weight:.2f} kg, " # Added weight
                f"stackable={placement.box.stackable}, "
                f"rotated={placement.rotated}"
            )

        # Removed print_3d_layers call as requested by the user.
        # print_3d_layers(
        #     main_box.width,
        #     main_box.depth,
        #     main_box.height,
        #     placements,
        #     main_box_name=(
        #         f"Main Box {result['usage_number']} "
        #         f"- {main_box.name}"
        #     )
        # )

    # --------------------------------------------------------
    # PRINT ITEMS THAT COULD NOT FIT ANY MAIN BOX
    # --------------------------------------------------------

    if remaining_boxes:
        print("\n" + "=" * 55)
        print("Item Boxes That Could Not Fit")
        print("=" * 55)

        # Group remaining boxes by type, dimensions, and weight
        grouped_unpacked = Counter()
        for box in remaining_boxes:
            # Remove numbering suffix (e.g., "-1", "-2")
            base_name = re.sub(r'-\d+$', '', box.name)
            key = (
                base_name,
                box.width,
                box.depth,
                box.height,
                box.weight,
                box.stackable
            )
            grouped_unpacked[key] += 1

        for (
            name,
            width,
            depth,
            height,
            weight,
            stackable
        ), count in grouped_unpacked.items():
            stack_label = "Stackable" if stackable else "Unstackable"
            print(
                f"- {name}: "
                f"{width:.1f} × {depth:.1f} × {height:.1f} cm, "
                f"Weight: {weight:.1f} kg, "
                f"Stacking: {stack_label}, "
                f"Amount: {count}"
            )

    # --------------------------------------------------------
    # DISPLAY ONE PLOTLY FIGURE PER USED MAIN BOX
    # --------------------------------------------------------

    visualize_multiple_main_boxes_3d(used_main_boxes_3d)

    # --------------------------------------------------------
    # OPTIONAL A4 PDF EXPORT
    # --------------------------------------------------------

    export_pdf = get_yes_no(
        "\nExport the complete result to an A4 PDF? yes/no: "
    )

    if export_pdf:
        try:
            pdf_path = create_packing_report_pdf(
                used_main_boxes_3d=used_main_boxes_3d,
                remaining_boxes=remaining_boxes,
                total_item_count=len(item_boxes_3d),
                packed_count=packed_count,
                item_gap=item_gap,
                container_gap=container_gap,
                center_items=center_items,
                output_filename="packing_report.pdf"
            )

            print(f"PDF report created successfully: {pdf_path}")
            print("The PDF contains static snapshots; the notebook plots remain interactive.")
            download_file_in_colab(pdf_path)

        except Exception as error:
            print("\nPDF export failed.")
            print(error)

    # --------------------------------------------------------
    # OPTIONAL INTERACTIVE HTML EXPORT
    # --------------------------------------------------------

    export_html = get_yes_no(
        "\nExport the complete result to an interactive HTML report? yes/no: "
    )

    if export_html:
        try:
            html_report_path = create_packing_report_html(
                used_main_boxes_3d=used_main_boxes_3d,
                remaining_boxes=remaining_boxes,
                total_item_count=len(item_boxes_3d),
                packed_count=packed_count,
                item_gap=item_gap,
                container_gap=container_gap,
                center_items=center_items,
                output_filename="packing_report.html"
            )
            print(f"Interactive HTML report created successfully: {html_report_path}")
            print("Open it in a browser to rotate, zoom, pan, and inspect the 3D views.")
            download_file_in_colab(html_report_path)
        except Exception as error:
            print("\nHTML export failed.")
            print(error)


# Run the program.
main_3d()
```
